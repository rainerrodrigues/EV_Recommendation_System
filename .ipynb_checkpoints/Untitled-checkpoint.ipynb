```python
def analyze_correlations(df, target_col='target_high_efficiency'):
    numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
    corr_matrix = df[numerical_cols].corr()
    
    target_corr = corr_matrix[target_col].drop(target_col).sort_values(ascending=False)
    
    print(f"\n🎯 Features Correlation with Target (sorted):")
    print(target_corr)
    
    # Identify highly correlated features with target
    high_corr_features = target_corr[abs(target_corr) > 0.3].index.tolist()
    print(f"\n✅ Highly correlated features (|corr| > 0.3):")
    for feat in high_corr_features:
        print(f"  - {feat}: {target_corr[feat]:.3f}")
    
    # Feature multicollinearity check
    print(f"\n🔍 Multicollinearity Check (highly correlated feature pairs):")
    multicollinear_pairs = []
    for i in range(len(corr_matrix.columns)):
        for j in range(i+1, len(corr_matrix.columns)):
            if abs(corr_matrix.iloc[i, j]) > 0.8:
                multicollinear_pairs.append(
                    (corr_matrix.columns[i], corr_matrix.columns[j], corr_matrix.iloc[i, j])
                )
    
    if multicollinear_pairs:
        print("High multicollinearity detected:")
        for feat1, feat2, corr_val in multicollinear_pairs:
            print(f"  - {feat1} <-> {feat2}: {corr_val:.3f}")
    else:
        print("No severe multicollinearity detected")
    
    # Plot correlation heatmap
    plt.figure(figsize=(14, 10))
    sns.heatmap(corr_matrix, annot=True, fmt='.2f', cmap='coolwarm', center=0,
                square=True, linewidths=1, cbar_kws={"shrink": 0.8})
    plt.title('Feature Correlation Matrix', fontsize=16, fontweight='bold')
    plt.tight_layout()
    plt.savefig('correlation_heatmap.png', dpi=300, bbox_inches='tight')
    print(f"\nCorrelation heatmap saved as 'correlation_heatmap.png'")
    plt.close()
    
    # Plot target correlation bar chart
    plt.figure(figsize=(10, 6))
    target_corr.plot(kind='barh', color=['green' if x > 0 else 'red' for x in target_corr])
    plt.xlabel('Correlation with Target', fontsize=12)
    plt.title('Feature Correlation with Target Variable', fontsize=14, fontweight='bold')
    plt.axvline(x=0, color='black', linestyle='--', linewidth=0.8)
    plt.tight_layout()
    plt.savefig('target_correlation.png', dpi=300, bbox_inches='tight')
    print(f"📊 Target correlation chart saved as 'target_correlation.png'")
    plt.close()
    
    return target_corr, high_corr_features, multicollinear_pairs
```

```python
target_corr, high_corr_features, multicollinear_pairs = analyze_correlations(df, numerical_features)
```

```python
def engineer_features(df):
    """Create derived features for better predictions"""
    
    print("\n" + "="*80)
    print("FEATURE ENGINEERING")
    print("="*80)
    
    df_eng = df.copy()
    
    # Range efficiency (km per kWh)
    if 'range_km' in df.columns and 'battery_kwh' in df.columns:
        df_eng['range_efficiency'] = df_eng['range_km'] / (df_eng['battery_kwh'] + 0.001)
        print("✅ Created: range_efficiency = range_km / battery_kwh")
    
    # Power-to-weight ratio proxy (using acceleration as inverse weight indicator)
    if 'acceleration_0_100_kmph' in df.columns:
        df_eng['power_proxy'] = 100 / (df_eng['acceleration_0_100_kmph'] + 0.001)
        print("✅ Created: power_proxy = 100 / acceleration_0_100_kmph")
    
    # Speed efficiency (range per top speed)
    if 'range_km' in df.columns and 'top_speed_kmph' in df.columns:
        df_eng['speed_efficiency'] = df_eng['range_km'] / (df_eng['top_speed_kmph'] + 0.001)
        print("✅ Created: speed_efficiency = range_km / top_speed_kmph")
    
    # Vehicle age
    if 'release_year' in df.columns:
        df_eng['vehicle_age'] = 2026 - df_eng['release_year']
        print("✅ Created: vehicle_age = 2026 - release_year")
    
    # Charging speed (battery per charging hour)
    if 'battery_kwh' in df.columns and 'charging_time_hr' in df.columns:
        df_eng['charging_speed'] = df_eng['battery_kwh'] / (df_eng['charging_time_hr'] + 0.001)
        print("✅ Created: charging_speed = battery_kwh / charging_time_hr")
    
    return df_eng
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    classification_report, 
    confusion_matrix, 
    roc_auc_score, 
    roc_curve,
    accuracy_score,
    precision_recall_curve
)
```

```python
def preprocess_data(df, numerical_features, categorical_features, 
                    target_col='target_high_efficiency', test_size=0.2, random_state=42):
    """Preprocess data for modeling"""
    
    print("\n" + "="*80)
    print("DATA PREPROCESSING")
    print("="*80)
    
    # Prepare feature matrix
    X = df[numerical_features + categorical_features].copy()
    y = df[target_col].copy()
    
    # Handle missing values
    print(f"\nHandling missing values...")
    for col in numerical_features:
        if X[col].isnull().sum() > 0:
            X[col].fillna(X[col].median(), inplace=True)
            print(f"  - Filled {col} with median")
    
    for col in categorical_features:
        if X[col].isnull().sum() > 0:
            X[col].fillna(X[col].mode()[0], inplace=True)
            print(f"  - Filled {col} with mode")
    
    # Encode categorical variables
    print(f"\nEncoding categorical variables...")
    label_encoders = {}
    for col in categorical_features:
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col].astype(str))
        label_encoders[col] = le
        print(f"  - Encoded {col} ({len(le.classes_)} classes)")
    
    # Train-test split
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )
    
    print(f"\n📊 Data Split:")
    print(f"  - Training set: {X_train.shape}")
    print(f"  - Test set: {X_test.shape}")
    print(f"  - Target distribution in train: {y_train.value_counts(normalize=True).to_dict()}")
    
    # Scale numerical features
    scaler = StandardScaler()
    X_train[numerical_features] = scaler.fit_transform(X_train[numerical_features])
    X_test[numerical_features] = scaler.transform(X_test[numerical_features])
    
    print(f"\n✅ Scaled numerical features using StandardScaler")
    
    return X_train, X_test, y_train, y_test, scaler, label_encoders
```

```python
X_train, X_test, y_train, y_test, scaler, label_encoders = preprocess_data(df, numerical_features, categorical_features)
```

```python
def train_baseline_model(X_train, X_test, y_train, y_test):
    """Train and evaluate baseline Logistic Regression model"""
    
    print("\n" + "="*80)
    print("BASELINE MODEL: LOGISTIC REGRESSION")
    print("="*80)
    
    # Train model
    print(f"\nTraining Logistic Regression...")
    model = LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced')
    model.fit(X_train, y_train)
    
    # Predictions
    y_pred_train = model.predict(X_train)
    y_pred_test = model.predict(X_test)
    y_pred_proba = model.predict_proba(X_test)[:, 1]
    
    # Evaluation
    print(f"\n📊 MODEL PERFORMANCE:")
    print(f"\n{'='*60}")
    print(f"TRAINING SET:")
    print(f"{'='*60}")
    print(f"Accuracy: {accuracy_score(y_train, y_pred_train):.4f}")
    
    print(f"\n{'='*60}")
    print(f"TEST SET:")
    print(f"{'='*60}")
    print(f"Accuracy: {accuracy_score(y_test, y_pred_test):.4f}")
    print(f"ROC-AUC Score: {roc_auc_score(y_test, y_pred_proba):.4f}")
    
    print(f"\n📋 Classification Report:")
    print(classification_report(y_test, y_pred_test, 
                                target_names=['Low Efficiency (0)', 'High Efficiency (1)']))
    
    # Confusion Matrix
    cm = confusion_matrix(y_test, y_pred_test)
    plt.figure(figsize=(8, 6))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
                xticklabels=['Low (0)', 'High (1)'],
                yticklabels=['Low (0)', 'High (1)'])
    plt.title('Confusion Matrix - Baseline Model', fontsize=14, fontweight='bold')
    plt.ylabel('Actual', fontsize=12)
    plt.xlabel('Predicted', fontsize=12)
    plt.tight_layout()
    plt.savefig('confusion_matrix_baseline.png', dpi=300, bbox_inches='tight')
    print(f"\n📊 Confusion matrix saved as 'confusion_matrix_baseline.png'")
    plt.close()
    
    # ROC Curve
    fpr, tpr, _ = roc_curve(y_test, y_pred_proba)
    plt.figure(figsize=(8, 6))
    plt.plot(fpr, tpr, linewidth=2, label=f'ROC Curve (AUC = {roc_auc_score(y_test, y_pred_proba):.3f})')
    plt.plot([0, 1], [0, 1], 'k--', linewidth=1, label='Random Classifier')
    plt.xlabel('False Positive Rate', fontsize=12)
    plt.ylabel('True Positive Rate', fontsize=12)
    plt.title('ROC Curve - Baseline Model', fontsize=14, fontweight='bold')
    plt.legend(fontsize=10)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig('roc_curve_baseline.png', dpi=300, bbox_inches='tight')
    print(f"📊 ROC curve saved as 'roc_curve_baseline.png'")
    plt.close()
    
    # Feature Importance
    feature_importance = pd.DataFrame({
        'feature': X_train.columns,
        'coefficient': model.coef_[0]
    }).sort_values('coefficient', key=abs, ascending=False)
    
    print(f"\n🎯 Top 10 Most Important Features (by coefficient magnitude):")
    print(feature_importance.head(10))
    
    # Plot feature importance
    plt.figure(figsize=(10, 8))
    top_features = feature_importance.head(15)
    colors = ['green' if x > 0 else 'red' for x in top_features['coefficient']]
    plt.barh(range(len(top_features)), top_features['coefficient'], color=colors)
    plt.yticks(range(len(top_features)), top_features['feature'])
    plt.xlabel('Coefficient Value', fontsize=12)
    plt.title('Top 15 Feature Importance (Logistic Regression)', fontsize=14, fontweight='bold')
    plt.axvline(x=0, color='black', linestyle='--', linewidth=0.8)
    plt.tight_layout()
    plt.savefig('feature_importance_baseline.png', dpi=300, bbox_inches='tight')
    print(f"📊 Feature importance chart saved as 'feature_importance_baseline.png'")
    plt.close()
    
    return model, feature_importance
```

```python
model, feature_importance = train_baseline_model(X_train, X_test, y_train, y_test)
```

```python
"""
CORRECTED Baseline Model - WITHOUT Data Leakage
================================================
This script properly handles the efficiency_score leakage issue
and uses engineered features for legitimate predictions.
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report, 
    confusion_matrix, 
    roc_auc_score, 
    roc_curve,
    accuracy_score
)
import warnings
warnings.filterwarnings('ignore')

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

# ============================================================================
# CORRECTED FEATURE ENGINEERING
# ============================================================================

def engineer_features_corrected(df):
    """
    Create derived features WITHOUT using efficiency_score
    These features are available BEFORE knowing the target
    """
    
    print("\n" + "="*80)
    print("CORRECTED FEATURE ENGINEERING (No Data Leakage)")
    print("="*80)
    
    df_eng = df.copy()
    
    # 1. Range efficiency (km per kWh) - MOST IMPORTANT
    if 'range_km' in df.columns and 'battery_kwh' in df.columns:
        df_eng['range_efficiency'] = df_eng['range_km'] / (df_eng['battery_kwh'] + 0.001)
        print("✅ Created: range_efficiency = range_km / battery_kwh")
    
    # 2. Battery to range ratio
    if 'battery_kwh' in df.columns and 'range_km' in df.columns:
        df_eng['battery_range_ratio'] = df_eng['battery_kwh'] / (df_eng['range_km'] + 0.001)
        print("✅ Created: battery_range_ratio = battery_kwh / range_km")
    
    # 3. Charging efficiency (battery per hour)
    if 'battery_kwh' in df.columns and 'charging_time_hr' in df.columns:
        df_eng['charging_speed'] = df_eng['battery_kwh'] / (df_eng['charging_time_hr'] + 0.001)
        print("✅ Created: charging_speed = battery_kwh / charging_time_hr")
    
    # 4. Performance proxy (inverse of acceleration)
    if 'acceleration_0_100_kmph' in df.columns:
        df_eng['power_indicator'] = 100 / (df_eng['acceleration_0_100_kmph'] + 0.001)
        print("✅ Created: power_indicator = 100 / acceleration_0_100_kmph")
    
    # 5. Speed efficiency
    if 'range_km' in df.columns and 'top_speed_kmph' in df.columns:
        df_eng['speed_efficiency'] = df_eng['range_km'] / (df_eng['top_speed_kmph'] + 0.001)
        print("✅ Created: speed_efficiency = range_km / top_speed_kmph")
    
    # 6. Vehicle age
    if 'release_year' in df.columns:
        df_eng['vehicle_age'] = 2026 - df_eng['release_year']
        print("✅ Created: vehicle_age = 2026 - release_year")
    
    # 7. Price per kWh (cost efficiency indicator)
    if 'price_usd' in df.columns and 'battery_kwh' in df.columns:
        df_eng['price_per_kwh'] = df_eng['price_usd'] / (df_eng['battery_kwh'] + 0.001)
        print("✅ Created: price_per_kwh = price_usd / battery_kwh")
    
    # 8. Range per seat (capacity efficiency)
    if 'range_km' in df.columns and 'seats' in df.columns:
        df_eng['range_per_seat'] = df_eng['range_km'] / (df_eng['seats'] + 0.001)
        print("✅ Created: range_per_seat = range_km / seats")
    
    # 9. Binary: Is Electric?
    if 'fuel_type' in df.columns:
        df_eng['is_electric'] = (df_eng['fuel_type'] == 'Electric').astype(int)
        print("✅ Created: is_electric (1 if Electric, 0 otherwise)")
    
    # 10. Binary: Has Fast Charging?
    if 'fast_charging' in df.columns:
        df_eng['has_fast_charging'] = df_eng['fast_charging'].astype(int)
        print("✅ Created: has_fast_charging (already binary)")
    
    return df_eng

# ============================================================================
# CORRECTED FEATURE SELECTION (Removes efficiency_score)
# ============================================================================

def select_features_corrected(df, target_col='target_high_efficiency'):
    """
    Select features WITHOUT data leakage
    
    Excludes:
    - efficiency_score (DATA LEAKAGE)
    - High cardinality categoricals (manufacturer, model, city)
    - Target variable
    """
    
    print("\n" + "="*80)
    print("CORRECTED FEATURE SELECTION")
    print("="*80)
    
    # Define exclusions
    leakage_features = ['efficiency_score']
    high_cardinality = ['manufacturer', 'model', 'city', 'color']  # color has 8 classes, might be noise
    
    # Get all features
    numerical_features = df.select_dtypes(include=[np.number]).columns.tolist()
    categorical_features = df.select_dtypes(include=['object']).columns.tolist()
    
    # Remove target
    if target_col in numerical_features:
        numerical_features.remove(target_col)
    
    # Remove data leakage features
    for feat in leakage_features:
        if feat in numerical_features:
            numerical_features.remove(feat)
            print(f"🚨 REMOVED '{feat}' - DATA LEAKAGE!")
    
    # Remove high cardinality
    categorical_features = [col for col in categorical_features 
                           if col not in high_cardinality]
    
    print(f"\n✅ Selected Features:")
    print(f"  - Numerical: {len(numerical_features)} features")
    print(f"    {numerical_features}")
    print(f"  - Categorical: {len(categorical_features)} features")
    print(f"    {categorical_features}")
    
    print(f"\n🚫 Excluded:")
    print(f"  - Data leakage: {leakage_features}")
    print(f"  - High cardinality: {high_cardinality}")
    
    return numerical_features, categorical_features

# ============================================================================
# ANALYZE CORRELATIONS (Without efficiency_score)
# ============================================================================

def analyze_correlations_corrected(df, numerical_cols, target_col='target_high_efficiency'):
    """Correlation analysis WITHOUT the leakage feature"""
    
    print("\n" + "="*80)
    print("CORRELATION ANALYSIS (Corrected)")
    print("="*80)
    
    # Remove efficiency_score if present
    numerical_cols_clean = [col for col in numerical_cols if col != 'efficiency_score']
    
    corr_matrix = df[numerical_cols_clean].corr()
    target_corr = corr_matrix[target_col].drop(target_col).sort_values(ascending=False)
    
    print(f"\n🎯 Feature Correlation with Target (WITHOUT efficiency_score):")
    print(target_corr)
    
    # Identify important features
    important_features = target_corr[abs(target_corr) > 0.1].index.tolist()
    if important_features:
        print(f"\n✅ Important features (|corr| > 0.1):")
        for feat in important_features:
            print(f"  - {feat}: {target_corr[feat]:.4f}")
    else:
        print(f"\n⚠️  No features with |correlation| > 0.1")
        print(f"   Top 5 features by correlation:")
        for feat in target_corr.head(5).index:
            print(f"  - {feat}: {target_corr[feat]:.4f}")
    
    # Plot
    plt.figure(figsize=(10, 8))
    top_15 = target_corr.head(15) if len(target_corr) > 15 else target_corr
    colors = ['green' if x > 0 else 'red' for x in top_15]
    plt.barh(range(len(top_15)), top_15, color=colors, alpha=0.7)
    plt.yticks(range(len(top_15)), top_15.index)
    plt.xlabel('Correlation with Target', fontsize=12)
    plt.title('Feature Correlation (Corrected - No Leakage)', fontsize=14, fontweight='bold')
    plt.axvline(x=0, color='black', linestyle='--', linewidth=0.8)
    plt.tight_layout()
    plt.savefig('target_correlation_corrected.png', dpi=300, bbox_inches='tight')
    print(f"\n📊 Saved: target_correlation_corrected.png")
    plt.close()
    
    return target_corr

# ============================================================================
# PREPROCESSING
# ============================================================================

def preprocess_data_corrected(df, numerical_features, categorical_features, 
                              target_col='target_high_efficiency', test_size=0.2, random_state=42):
    """Preprocess data for modeling"""
    
    print("\n" + "="*80)
    print("DATA PREPROCESSING")
    print("="*80)
    
    X = df[numerical_features + categorical_features].copy()
    y = df[target_col].copy()
    
    # Handle missing values
    for col in numerical_features:
        if X[col].isnull().sum() > 0:
            X[col].fillna(X[col].median(), inplace=True)
    
    for col in categorical_features:
        if X[col].isnull().sum() > 0:
            X[col].fillna(X[col].mode()[0], inplace=True)
    
    # Encode categoricals
    label_encoders = {}
    for col in categorical_features:
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col].astype(str))
        label_encoders[col] = le
    
    # Split
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )
    
    print(f"\n📊 Data Split:")
    print(f"  - Training: {X_train.shape}")
    print(f"  - Test: {X_test.shape}")
    print(f"  - Class balance: {y_train.value_counts(normalize=True).to_dict()}")
    
    # Scale numerical features
    scaler = StandardScaler()
    X_train[numerical_features] = scaler.fit_transform(X_train[numerical_features])
    X_test[numerical_features] = scaler.transform(X_test[numerical_features])
    
    print(f"✅ Scaled numerical features")
    
    return X_train, X_test, y_train, y_test, scaler, label_encoders

# ============================================================================
# TRAIN MODELS (Logistic Regression + Random Forest)
# ============================================================================

def train_models_corrected(X_train, X_test, y_train, y_test):
    """Train both baseline and improved models"""
    
    print("\n" + "="*80)
    print("MODEL TRAINING (Corrected - No Data Leakage)")
    print("="*80)
    
    results = {}
    
    # ========== MODEL 1: Logistic Regression ==========
    print(f"\n{'='*60}")
    print(f"MODEL 1: LOGISTIC REGRESSION (Baseline)")
    print(f"{'='*60}")
    
    lr_model = LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced')
    lr_model.fit(X_train, y_train)
    
    lr_pred_train = lr_model.predict(X_train)
    lr_pred_test = lr_model.predict(X_test)
    lr_pred_proba = lr_model.predict_proba(X_test)[:, 1]
    
    print(f"\n📊 Logistic Regression Performance:")
    print(f"  Training Accuracy: {accuracy_score(y_train, lr_pred_train):.4f}")
    print(f"  Test Accuracy: {accuracy_score(y_test, lr_pred_test):.4f}")
    print(f"  ROC-AUC: {roc_auc_score(y_test, lr_pred_proba):.4f}")
    
    # Cross-validation
    cv_scores = cross_val_score(lr_model, X_train, y_train, cv=5, scoring='roc_auc')
    print(f"  CV ROC-AUC (5-fold): {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})")
    
    print(f"\n📋 Classification Report:")
    print(classification_report(y_test, lr_pred_test))
    
    results['logistic_regression'] = {
        'model': lr_model,
        'accuracy': accuracy_score(y_test, lr_pred_test),
        'roc_auc': roc_auc_score(y_test, lr_pred_proba),
        'predictions': lr_pred_test,
        'probabilities': lr_pred_proba
    }
    
    # Feature importance
    lr_importance = pd.DataFrame({
        'feature': X_train.columns,
        'coefficient': lr_model.coef_[0]
    }).sort_values('coefficient', key=abs, ascending=False)
    
    print(f"\n🎯 Top 10 Features (Logistic Regression):")
    print(lr_importance.head(10))
    
    # ========== MODEL 2: Random Forest ==========
    print(f"\n{'='*60}")
    print(f"MODEL 2: RANDOM FOREST")
    print(f"{'='*60}")
    
    rf_model = RandomForestClassifier(
        n_estimators=100, 
        max_depth=10,
        min_samples_split=100,
        min_samples_leaf=50,
        random_state=42,
        class_weight='balanced',
        n_jobs=-1
    )
    rf_model.fit(X_train, y_train)
    
    rf_pred_train = rf_model.predict(X_train)
    rf_pred_test = rf_model.predict(X_test)
    rf_pred_proba = rf_model.predict_proba(X_test)[:, 1]
    
    print(f"\n📊 Random Forest Performance:")
    print(f"  Training Accuracy: {accuracy_score(y_train, rf_pred_train):.4f}")
    print(f"  Test Accuracy: {accuracy_score(y_test, rf_pred_test):.4f}")
    print(f"  ROC-AUC: {roc_auc_score(y_test, rf_pred_proba):.4f}")
    
    # Cross-validation
    cv_scores_rf = cross_val_score(rf_model, X_train, y_train, cv=5, scoring='roc_auc')
    print(f"  CV ROC-AUC (5-fold): {cv_scores_rf.mean():.4f} (+/- {cv_scores_rf.std():.4f})")
    
    print(f"\n📋 Classification Report:")
    print(classification_report(y_test, rf_pred_test))
    
    results['random_forest'] = {
        'model': rf_model,
        'accuracy': accuracy_score(y_test, rf_pred_test),
        'roc_auc': roc_auc_score(y_test, rf_pred_proba),
        'predictions': rf_pred_test,
        'probabilities': rf_pred_proba
    }
    
    # Feature importance
    rf_importance = pd.DataFrame({
        'feature': X_train.columns,
        'importance': rf_model.feature_importances_
    }).sort_values('importance', ascending=False)
    
    print(f"\n🎯 Top 10 Features (Random Forest):")
    print(rf_importance.head(10))
    
    # ========== VISUALIZATIONS ==========
    
    # Confusion matrices
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    for idx, (name, result) in enumerate([('Logistic Regression', results['logistic_regression']),
                                          ('Random Forest', results['random_forest'])]):
        cm = confusion_matrix(y_test, result['predictions'])
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[idx],
                   xticklabels=['Low', 'High'], yticklabels=['Low', 'High'])
        axes[idx].set_title(f'{name}\nAccuracy: {result["accuracy"]:.4f}', fontweight='bold')
        axes[idx].set_ylabel('Actual')
        axes[idx].set_xlabel('Predicted')
    
    plt.tight_layout()
    plt.savefig('confusion_matrices_corrected.png', dpi=300, bbox_inches='tight')
    print(f"\n📊 Saved: confusion_matrices_corrected.png")
    plt.close()
    
    # ROC curves
    plt.figure(figsize=(10, 6))
    
    for name, result in results.items():
        fpr, tpr, _ = roc_curve(y_test, result['probabilities'])
        plt.plot(fpr, tpr, linewidth=2, 
                label=f'{name.replace("_", " ").title()} (AUC={result["roc_auc"]:.3f})')
    
    plt.plot([0, 1], [0, 1], 'k--', linewidth=1, label='Random')
    plt.xlabel('False Positive Rate', fontsize=12)
    plt.ylabel('True Positive Rate', fontsize=12)
    plt.title('ROC Curves (Corrected Models)', fontsize=14, fontweight='bold')
    plt.legend(fontsize=10)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig('roc_curves_corrected.png', dpi=300, bbox_inches='tight')
    print(f"📊 Saved: roc_curves_corrected.png")
    plt.close()
    
    # Feature importance comparison
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    
    # Logistic Regression
    top_lr = lr_importance.head(10)
    colors_lr = ['green' if x > 0 else 'red' for x in top_lr['coefficient']]
    axes[0].barh(range(len(top_lr)), top_lr['coefficient'], color=colors_lr, alpha=0.7)
    axes[0].set_yticks(range(len(top_lr)))
    axes[0].set_yticklabels(top_lr['feature'])
    axes[0].set_xlabel('Coefficient')
    axes[0].set_title('Logistic Regression Feature Importance', fontweight='bold')
    axes[0].axvline(x=0, color='black', linestyle='--', linewidth=0.8)
    
    # Random Forest
    top_rf = rf_importance.head(10)
    axes[1].barh(range(len(top_rf)), top_rf['importance'], color='steelblue', alpha=0.7)
    axes[1].set_yticks(range(len(top_rf)))
    axes[1].set_yticklabels(top_rf['feature'])
    axes[1].set_xlabel('Importance')
    axes[1].set_title('Random Forest Feature Importance', fontweight='bold')
    
    plt.tight_layout()
    plt.savefig('feature_importance_corrected.png', dpi=300, bbox_inches='tight')
    print(f"📊 Saved: feature_importance_corrected.png")
    plt.close()
    
    return results, lr_importance, rf_importance

# ============================================================================
# MAIN EXECUTION
# ============================================================================

def main(filepath):
    """Main execution pipeline - CORRECTED VERSION"""
    
    print("\n" + "="*80)
    print("🚗 EV EFFICIENCY PREDICTION - CORRECTED PIPELINE")
    print("   (No Data Leakage - Using Engineered Features)")
    print("="*80)
    
    # Load data
    print(f"\nLoading data from: {filepath}")
    df = pd.read_csv(filepath)
    print(f"Dataset shape: {df.shape}")
    
    # Feature engineering
    df_eng = engineer_features_corrected(df)
    
    # Feature selection (removes efficiency_score)
    numerical_features, categorical_features = select_features_corrected(df_eng)
    
    # Correlation analysis
    target_corr = analyze_correlations_corrected(df_eng, numerical_features + [df_eng.columns[df_eng.columns == 'target_high_efficiency'][0]])
    
    # Preprocessing
    X_train, X_test, y_train, y_test, scaler, label_encoders = preprocess_data_corrected(
        df_eng, numerical_features, categorical_features
    )
    
    # Train models
    results, lr_importance, rf_importance = train_models_corrected(
        X_train, X_test, y_train, y_test
    )
    
    # Summary
    print("\n" + "="*80)
    print("✅ PIPELINE COMPLETED!")
    print("="*80)
    
    print(f"\n📊 Model Comparison:")
    print(f"{'Model':<25} {'Accuracy':<12} {'ROC-AUC':<12}")
    print(f"{'-'*50}")
    for name, result in results.items():
        print(f"{name.replace('_', ' ').title():<25} {result['accuracy']:<12.4f} {result['roc_auc']:<12.4f}")
    
    print(f"\n📁 Generated Files:")
    print(f"  - target_correlation_corrected.png")
    print(f"  - confusion_matrices_corrected.png")
    print(f"  - roc_curves_corrected.png")
    print(f"  - feature_importance_corrected.png")
    
    print(f"\n💡 Next Steps:")
    if results['logistic_regression']['roc_auc'] < 0.70:
        print(f"  ⚠️  Performance is lower than expected (ROC-AUC < 0.70)")
        print(f"  → This is NORMAL after removing data leakage!")
        print(f"  → Your target might be genuinely hard to predict from these features")
        print(f"  → Consider:")
        print(f"     - Collect more relevant features (vehicle weight, motor power, etc.)")
        print(f"     - Try XGBoost/LightGBM for better performance")
        print(f"     - Check if target definition makes sense")
    else:
        print(f"  ✅ Good performance! Now:")
        print(f"     1. Try XGBoost/LightGBM")
        print(f"     2. Hyperparameter tuning")
        print(f"     3. Ensemble methods")
    
    return df_eng, results, lr_importance, rf_importance

if __name__ == "__main__":
    FILEPATH = "dataset/cleaned_dataset.csv"  # <-- CHANGE THIS
    
    df_eng, results, lr_imp, rf_imp = main(FILEPATH)
```

```python
"""
CORRECTED Baseline Model - WITHOUT Data Leakage
================================================
This script properly handles the efficiency_score leakage issue
and uses engineered features for legitimate predictions.
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report, 
    confusion_matrix, 
    roc_auc_score, 
    roc_curve,
    accuracy_score
)
import warnings
warnings.filterwarnings('ignore')

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

# ============================================================================
# CORRECTED FEATURE ENGINEERING
# ============================================================================

def engineer_features_corrected(df):
    """
    Create derived features WITHOUT using efficiency_score
    These features are available BEFORE knowing the target
    """
    
    print("\n" + "="*80)
    print("CORRECTED FEATURE ENGINEERING (No Data Leakage)")
    print("="*80)
    
    df_eng = df.copy()
    
    # 1. Range efficiency (km per kWh) - MOST IMPORTANT
    if 'range_km' in df.columns and 'battery_kwh' in df.columns:
        df_eng['range_efficiency'] = df_eng['range_km'] / (df_eng['battery_kwh'] + 0.001)
        print("✅ Created: range_efficiency = range_km / battery_kwh")
    
    # 2. Battery to range ratio
    if 'battery_kwh' in df.columns and 'range_km' in df.columns:
        df_eng['battery_range_ratio'] = df_eng['battery_kwh'] / (df_eng['range_km'] + 0.001)
        print("✅ Created: battery_range_ratio = battery_kwh / range_km")
    
    # 3. Charging efficiency (battery per hour)
    if 'battery_kwh' in df.columns and 'charging_time_hr' in df.columns:
        df_eng['charging_speed'] = df_eng['battery_kwh'] / (df_eng['charging_time_hr'] + 0.001)
        print("✅ Created: charging_speed = battery_kwh / charging_time_hr")
    
    # 4. Performance proxy (inverse of acceleration)
    if 'acceleration_0_100_kmph' in df.columns:
        df_eng['power_indicator'] = 100 / (df_eng['acceleration_0_100_kmph'] + 0.001)
        print("✅ Created: power_indicator = 100 / acceleration_0_100_kmph")
    
    # 5. Speed efficiency
    if 'range_km' in df.columns and 'top_speed_kmph' in df.columns:
        df_eng['speed_efficiency'] = df_eng['range_km'] / (df_eng['top_speed_kmph'] + 0.001)
        print("✅ Created: speed_efficiency = range_km / top_speed_kmph")
    
    # 6. Vehicle age
    if 'release_year' in df.columns:
        df_eng['vehicle_age'] = 2026 - df_eng['release_year']
        print("✅ Created: vehicle_age = 2026 - release_year")
    
    # 7. Price per kWh (cost efficiency indicator)
    if 'price_usd' in df.columns and 'battery_kwh' in df.columns:
        df_eng['price_per_kwh'] = df_eng['price_usd'] / (df_eng['battery_kwh'] + 0.001)
        print("✅ Created: price_per_kwh = price_usd / battery_kwh")
    
    # 8. Range per seat (capacity efficiency)
    if 'range_km' in df.columns and 'seats' in df.columns:
        df_eng['range_per_seat'] = df_eng['range_km'] / (df_eng['seats'] + 0.001)
        print("✅ Created: range_per_seat = range_km / seats")
    
    # 9. Binary: Is Electric?
    if 'fuel_type' in df.columns:
        df_eng['is_electric'] = (df_eng['fuel_type'] == 'Electric').astype(int)
        print("✅ Created: is_electric (1 if Electric, 0 otherwise)")
    
    # 10. Binary: Has Fast Charging?
    if 'fast_charging' in df.columns:
        df_eng['has_fast_charging'] = df_eng['fast_charging'].astype(int)
        print("✅ Created: has_fast_charging (already binary)")
    
    return df_eng

# ============================================================================
# CORRECTED FEATURE SELECTION (Removes efficiency_score)
# ============================================================================

def select_features_corrected(df, target_col='target_high_efficiency'):
    """
    Select features WITHOUT data leakage
    
    Excludes:
    - efficiency_score (DATA LEAKAGE)
    - High cardinality categoricals (manufacturer, model, city)
    - Target variable
    """
    
    print("\n" + "="*80)
    print("CORRECTED FEATURE SELECTION")
    print("="*80)
    
    # Define exclusions
    leakage_features = ['efficiency_score']
    high_cardinality = ['manufacturer', 'model', 'city', 'color']  # color has 8 classes, might be noise
    
    # Get all features
    numerical_features = df.select_dtypes(include=[np.number]).columns.tolist()
    categorical_features = df.select_dtypes(include=['object']).columns.tolist()
    
    # Remove target
    if target_col in numerical_features:
        numerical_features.remove(target_col)
    
    # Remove data leakage features
    for feat in leakage_features:
        if feat in numerical_features:
            numerical_features.remove(feat)
            print(f"🚨 REMOVED '{feat}' - DATA LEAKAGE!")
    
    # Remove high cardinality
    categorical_features = [col for col in categorical_features 
                           if col not in high_cardinality]
    
    print(f"\n✅ Selected Features:")
    print(f"  - Numerical: {len(numerical_features)} features")
    print(f"    {numerical_features}")
    print(f"  - Categorical: {len(categorical_features)} features")
    print(f"    {categorical_features}")
    
    print(f"\n🚫 Excluded:")
    print(f"  - Data leakage: {leakage_features}")
    print(f"  - High cardinality: {high_cardinality}")
    
    return numerical_features, categorical_features

# ============================================================================
# ANALYZE CORRELATIONS (Without efficiency_score)
# ============================================================================

def analyze_correlations_corrected(df, numerical_cols, target_col='target_high_efficiency'):
    """Correlation analysis WITHOUT the leakage feature"""
    
    print("\n" + "="*80)
    print("CORRELATION ANALYSIS (Corrected)")
    print("="*80)
    
    # Remove efficiency_score if present
    numerical_cols_clean = [col for col in numerical_cols if col != 'efficiency_score']
    
    corr_matrix = df[numerical_cols_clean].corr()
    target_corr = corr_matrix[target_col].drop(target_col).sort_values(ascending=False)
    
    print(f"\n🎯 Feature Correlation with Target (WITHOUT efficiency_score):")
    print(target_corr)
    
    # Identify important features
    important_features = target_corr[abs(target_corr) > 0.1].index.tolist()
    if important_features:
        print(f"\n✅ Important features (|corr| > 0.1):")
        for feat in important_features:
            print(f"  - {feat}: {target_corr[feat]:.4f}")
    else:
        print(f"\n⚠️  No features with |correlation| > 0.1")
        print(f"   Top 5 features by correlation:")
        for feat in target_corr.head(5).index:
            print(f"  - {feat}: {target_corr[feat]:.4f}")
    
    # Plot
    plt.figure(figsize=(10, 8))
    top_15 = target_corr.head(15) if len(target_corr) > 15 else target_corr
    colors = ['green' if x > 0 else 'red' for x in top_15]
    plt.barh(range(len(top_15)), top_15, color=colors, alpha=0.7)
    plt.yticks(range(len(top_15)), top_15.index)
    plt.xlabel('Correlation with Target', fontsize=12)
    plt.title('Feature Correlation (Corrected - No Leakage)', fontsize=14, fontweight='bold')
    plt.axvline(x=0, color='black', linestyle='--', linewidth=0.8)
    plt.tight_layout()
    plt.savefig('target_correlation_corrected.png', dpi=300, bbox_inches='tight')
    print(f"\n📊 Saved: target_correlation_corrected.png")
    plt.close()
    
    return target_corr

# ============================================================================
# PREPROCESSING
# ============================================================================

def preprocess_data_corrected(df, numerical_features, categorical_features, 
                              target_col='target_high_efficiency', test_size=0.2, random_state=42):
    """Preprocess data for modeling"""
    
    print("\n" + "="*80)
    print("DATA PREPROCESSING")
    print("="*80)
    
    X = df[numerical_features + categorical_features].copy()
    y = df[target_col].copy()
    
    # Handle missing values
    for col in numerical_features:
        if X[col].isnull().sum() > 0:
            X[col].fillna(X[col].median(), inplace=True)
    
    for col in categorical_features:
        if X[col].isnull().sum() > 0:
            X[col].fillna(X[col].mode()[0], inplace=True)
    
    # Encode categoricals
    label_encoders = {}
    for col in categorical_features:
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col].astype(str))
        label_encoders[col] = le
    
    # Split
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )
    
    print(f"\n📊 Data Split:")
    print(f"  - Training: {X_train.shape}")
    print(f"  - Test: {X_test.shape}")
    print(f"  - Class balance: {y_train.value_counts(normalize=True).to_dict()}")
    
    # Scale numerical features
    scaler = StandardScaler()
    X_train[numerical_features] = scaler.fit_transform(X_train[numerical_features])
    X_test[numerical_features] = scaler.transform(X_test[numerical_features])
    
    print(f"✅ Scaled numerical features")
    
    return X_train, X_test, y_train, y_test, scaler, label_encoders

# ============================================================================
# TRAIN MODELS (Logistic Regression + Random Forest)
# ============================================================================

def train_models_corrected(X_train, X_test, y_train, y_test):
    """Train both baseline and improved models"""
    
    print("\n" + "="*80)
    print("MODEL TRAINING (Corrected - No Data Leakage)")
    print("="*80)
    
    results = {}
    
    # ========== MODEL 1: Logistic Regression ==========
    print(f"\n{'='*60}")
    print(f"MODEL 1: LOGISTIC REGRESSION (Baseline)")
    print(f"{'='*60}")
    
    lr_model = LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced')
    lr_model.fit(X_train, y_train)
    
    lr_pred_train = lr_model.predict(X_train)
    lr_pred_test = lr_model.predict(X_test)
    lr_pred_proba = lr_model.predict_proba(X_test)[:, 1]
    
    print(f"\n📊 Logistic Regression Performance:")
    print(f"  Training Accuracy: {accuracy_score(y_train, lr_pred_train):.4f}")
    print(f"  Test Accuracy: {accuracy_score(y_test, lr_pred_test):.4f}")
    print(f"  ROC-AUC: {roc_auc_score(y_test, lr_pred_proba):.4f}")
    
    # Cross-validation
    cv_scores = cross_val_score(lr_model, X_train, y_train, cv=5, scoring='roc_auc')
    print(f"  CV ROC-AUC (5-fold): {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})")
    
    print(f"\n📋 Classification Report:")
    print(classification_report(y_test, lr_pred_test))
    
    results['logistic_regression'] = {
        'model': lr_model,
        'accuracy': accuracy_score(y_test, lr_pred_test),
        'roc_auc': roc_auc_score(y_test, lr_pred_proba),
        'predictions': lr_pred_test,
        'probabilities': lr_pred_proba
    }
    
    # Feature importance
    lr_importance = pd.DataFrame({
        'feature': X_train.columns,
        'coefficient': lr_model.coef_[0]
    }).sort_values('coefficient', key=abs, ascending=False)
    
    print(f"\n🎯 Top 10 Features (Logistic Regression):")
    print(lr_importance.head(10))
    
    # ========== MODEL 2: Random Forest ==========
    print(f"\n{'='*60}")
    print(f"MODEL 2: RANDOM FOREST")
    print(f"{'='*60}")
    
    rf_model = RandomForestClassifier(
        n_estimators=100, 
        max_depth=10,
        min_samples_split=100,
        min_samples_leaf=50,
        random_state=42,
        class_weight='balanced',
        n_jobs=-1
    )
    rf_model.fit(X_train, y_train)
    
    rf_pred_train = rf_model.predict(X_train)
    rf_pred_test = rf_model.predict(X_test)
    rf_pred_proba = rf_model.predict_proba(X_test)[:, 1]
    
    print(f"\n📊 Random Forest Performance:")
    print(f"  Training Accuracy: {accuracy_score(y_train, rf_pred_train):.4f}")
    print(f"  Test Accuracy: {accuracy_score(y_test, rf_pred_test):.4f}")
    print(f"  ROC-AUC: {roc_auc_score(y_test, rf_pred_proba):.4f}")
    
    # Cross-validation
    cv_scores_rf = cross_val_score(rf_model, X_train, y_train, cv=5, scoring='roc_auc')
    print(f"  CV ROC-AUC (5-fold): {cv_scores_rf.mean():.4f} (+/- {cv_scores_rf.std():.4f})")
    
    print(f"\n📋 Classification Report:")
    print(classification_report(y_test, rf_pred_test))
    
    results['random_forest'] = {
        'model': rf_model,
        'accuracy': accuracy_score(y_test, rf_pred_test),
        'roc_auc': roc_auc_score(y_test, rf_pred_proba),
        'predictions': rf_pred_test,
        'probabilities': rf_pred_proba
    }
    
    # Feature importance
    rf_importance = pd.DataFrame({
        'feature': X_train.columns,
        'importance': rf_model.feature_importances_
    }).sort_values('importance', ascending=False)
    
    print(f"\n🎯 Top 10 Features (Random Forest):")
    print(rf_importance.head(10))
    
    # ========== VISUALIZATIONS ==========
    
    # Confusion matrices
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    for idx, (name, result) in enumerate([('Logistic Regression', results['logistic_regression']),
                                          ('Random Forest', results['random_forest'])]):
        cm = confusion_matrix(y_test, result['predictions'])
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[idx],
                   xticklabels=['Low', 'High'], yticklabels=['Low', 'High'])
        axes[idx].set_title(f'{name}\nAccuracy: {result["accuracy"]:.4f}', fontweight='bold')
        axes[idx].set_ylabel('Actual')
        axes[idx].set_xlabel('Predicted')
    
    plt.tight_layout()
    plt.savefig('confusion_matrices_corrected.png', dpi=300, bbox_inches='tight')
    print(f"\n📊 Saved: confusion_matrices_corrected.png")
    plt.close()
    
    # ROC curves
    plt.figure(figsize=(10, 6))
    
    for name, result in results.items():
        fpr, tpr, _ = roc_curve(y_test, result['probabilities'])
        plt.plot(fpr, tpr, linewidth=2, 
                label=f'{name.replace("_", " ").title()} (AUC={result["roc_auc"]:.3f})')
    
    plt.plot([0, 1], [0, 1], 'k--', linewidth=1, label='Random')
    plt.xlabel('False Positive Rate', fontsize=12)
    plt.ylabel('True Positive Rate', fontsize=12)
    plt.title('ROC Curves (Corrected Models)', fontsize=14, fontweight='bold')
    plt.legend(fontsize=10)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig('roc_curves_corrected.png', dpi=300, bbox_inches='tight')
    print(f"📊 Saved: roc_curves_corrected.png")
    plt.close()
    
    # Feature importance comparison
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    
    # Logistic Regression
    top_lr = lr_importance.head(10)
    colors_lr = ['green' if x > 0 else 'red' for x in top_lr['coefficient']]
    axes[0].barh(range(len(top_lr)), top_lr['coefficient'], color=colors_lr, alpha=0.7)
    axes[0].set_yticks(range(len(top_lr)))
    axes[0].set_yticklabels(top_lr['feature'])
    axes[0].set_xlabel('Coefficient')
    axes[0].set_title('Logistic Regression Feature Importance', fontweight='bold')
    axes[0].axvline(x=0, color='black', linestyle='--', linewidth=0.8)
    
    # Random Forest
    top_rf = rf_importance.head(10)
    axes[1].barh(range(len(top_rf)), top_rf['importance'], color='steelblue', alpha=0.7)
    axes[1].set_yticks(range(len(top_rf)))
    axes[1].set_yticklabels(top_rf['feature'])
    axes[1].set_xlabel('Importance')
    axes[1].set_title('Random Forest Feature Importance', fontweight='bold')
    
    plt.tight_layout()
    plt.savefig('feature_importance_corrected.png', dpi=300, bbox_inches='tight')
    print(f"📊 Saved: feature_importance_corrected.png")
    plt.close()
    
    return results, lr_importance, rf_importance

# ============================================================================
# MAIN EXECUTION
# ============================================================================

def main(filepath):
    """Main execution pipeline - CORRECTED VERSION"""
    
    print("\n" + "="*80)
    print("🚗 EV EFFICIENCY PREDICTION - CORRECTED PIPELINE")
    print("   (No Data Leakage - Using Engineered Features)")
    print("="*80)
    
    # Load data
    print(f"\nLoading data from: {filepath}")
    df = pd.read_csv(filepath)
    print(f"Dataset shape: {df.shape}")
    
    # Feature engineering
    # df_eng = engineer_features_corrected(df)
    
    # Feature selection (removes efficiency_score)
    numerical_features, categorical_features = select_features_corrected(df)
    
    # Correlation analysis
    target_corr = analyze_correlations_corrected(df_eng, numerical_features + [df_eng.columns[df_eng.columns == 'target_high_efficiency'][0]])
    
    # Preprocessing
    X_train, X_test, y_train, y_test, scaler, label_encoders = preprocess_data_corrected(
        df_eng, numerical_features, categorical_features
    )
    
    # Train models
    results, lr_importance, rf_importance = train_models_corrected(
        X_train, X_test, y_train, y_test
    )
    
    # Summary
    print("\n" + "="*80)
    print("✅ PIPELINE COMPLETED!")
    print("="*80)
    
    print(f"\n📊 Model Comparison:")
    print(f"{'Model':<25} {'Accuracy':<12} {'ROC-AUC':<12}")
    print(f"{'-'*50}")
    for name, result in results.items():
        print(f"{name.replace('_', ' ').title():<25} {result['accuracy']:<12.4f} {result['roc_auc']:<12.4f}")
    
    print(f"\n📁 Generated Files:")
    print(f"  - target_correlation_corrected.png")
    print(f"  - confusion_matrices_corrected.png")
    print(f"  - roc_curves_corrected.png")
    print(f"  - feature_importance_corrected.png")
    
    print(f"\n💡 Next Steps:")
    if results['logistic_regression']['roc_auc'] < 0.70:
        print(f"  ⚠️  Performance is lower than expected (ROC-AUC < 0.70)")
        print(f"  → This is NORMAL after removing data leakage!")
        print(f"  → Your target might be genuinely hard to predict from these features")
        print(f"  → Consider:")
        print(f"     - Collect more relevant features (vehicle weight, motor power, etc.)")
        print(f"     - Try XGBoost/LightGBM for better performance")
        print(f"     - Check if target definition makes sense")
    else:
        print(f"  ✅ Good performance! Now:")
        print(f"     1. Try XGBoost/LightGBM")
        print(f"     2. Hyperparameter tuning")
        print(f"     3. Ensemble methods")
    
    return df_eng, results, lr_importance, rf_importance

if __name__ == "__main__":
    FILEPATH = "dataset/cleaned_dataset.csv"  # <-- CHANGE THIS
    
    df_eng, results, lr_imp, rf_imp = main(FILEPATH)
```

```python
"""
CORRECTED Baseline Model - WITHOUT Data Leakage
================================================
This script properly handles the efficiency_score leakage issue
and uses engineered features for legitimate predictions.
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report, 
    confusion_matrix, 
    roc_auc_score, 
    roc_curve,
    accuracy_score
)
import warnings
warnings.filterwarnings('ignore')

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

# ============================================================================
# CORRECTED FEATURE ENGINEERING
# ============================================================================

def engineer_features_corrected(df):
    """
    Create derived features WITHOUT using efficiency_score
    These features are available BEFORE knowing the target
    """
    
    print("\n" + "="*80)
    print("CORRECTED FEATURE ENGINEERING (No Data Leakage)")
    print("="*80)
    
    df_eng = df.copy()
    
    # 1. Range efficiency (km per kWh) - MOST IMPORTANT
    if 'range_km' in df.columns and 'battery_kwh' in df.columns:
        df_eng['range_efficiency'] = df_eng['range_km'] / (df_eng['battery_kwh'] + 0.001)
        print("✅ Created: range_efficiency = range_km / battery_kwh")
    
    # 2. Battery to range ratio
    if 'battery_kwh' in df.columns and 'range_km' in df.columns:
        df_eng['battery_range_ratio'] = df_eng['battery_kwh'] / (df_eng['range_km'] + 0.001)
        print("✅ Created: battery_range_ratio = battery_kwh / range_km")
    
    # 3. Charging efficiency (battery per hour)
    if 'battery_kwh' in df.columns and 'charging_time_hr' in df.columns:
        df_eng['charging_speed'] = df_eng['battery_kwh'] / (df_eng['charging_time_hr'] + 0.001)
        print("✅ Created: charging_speed = battery_kwh / charging_time_hr")
    
    # 4. Performance proxy (inverse of acceleration)
    if 'acceleration_0_100_kmph' in df.columns:
        df_eng['power_indicator'] = 100 / (df_eng['acceleration_0_100_kmph'] + 0.001)
        print("✅ Created: power_indicator = 100 / acceleration_0_100_kmph")
    
    # 5. Speed efficiency
    if 'range_km' in df.columns and 'top_speed_kmph' in df.columns:
        df_eng['speed_efficiency'] = df_eng['range_km'] / (df_eng['top_speed_kmph'] + 0.001)
        print("✅ Created: speed_efficiency = range_km / top_speed_kmph")
    
    # 6. Vehicle age
    if 'release_year' in df.columns:
        df_eng['vehicle_age'] = 2026 - df_eng['release_year']
        print("✅ Created: vehicle_age = 2026 - release_year")
    
    # 7. Price per kWh (cost efficiency indicator)
    if 'price_usd' in df.columns and 'battery_kwh' in df.columns:
        df_eng['price_per_kwh'] = df_eng['price_usd'] / (df_eng['battery_kwh'] + 0.001)
        print("✅ Created: price_per_kwh = price_usd / battery_kwh")
    
    # 8. Range per seat (capacity efficiency)
    if 'range_km' in df.columns and 'seats' in df.columns:
        df_eng['range_per_seat'] = df_eng['range_km'] / (df_eng['seats'] + 0.001)
        print("✅ Created: range_per_seat = range_km / seats")
    
    # 9. Binary: Is Electric?
    if 'fuel_type' in df.columns:
        df_eng['is_electric'] = (df_eng['fuel_type'] == 'Electric').astype(int)
        print("✅ Created: is_electric (1 if Electric, 0 otherwise)")
    
    # 10. Binary: Has Fast Charging?
    if 'fast_charging' in df.columns:
        df_eng['has_fast_charging'] = df_eng['fast_charging'].astype(int)
        print("✅ Created: has_fast_charging (already binary)")
    
    return df_eng

# ============================================================================
# CORRECTED FEATURE SELECTION (Removes efficiency_score)
# ============================================================================

def select_features_corrected(df, target_col='target_high_efficiency'):
    """
    Select features WITHOUT data leakage
    
    Excludes:
    - efficiency_score (DATA LEAKAGE)
    - High cardinality categoricals (manufacturer, model, city)
    - Target variable
    """
    
    print("\n" + "="*80)
    print("CORRECTED FEATURE SELECTION")
    print("="*80)
    
    # Define exclusions
    leakage_features = ['efficiency_score']
    high_cardinality = ['manufacturer', 'model', 'city', 'color']  # color has 8 classes, might be noise
    
    # Get all features
    numerical_features = df.select_dtypes(include=[np.number]).columns.tolist()
    categorical_features = df.select_dtypes(include=['object']).columns.tolist()
    
    # Remove target
    if target_col in numerical_features:
        numerical_features.remove(target_col)
    
    # Remove data leakage features
    for feat in leakage_features:
        if feat in numerical_features:
            numerical_features.remove(feat)
            print(f"🚨 REMOVED '{feat}' - DATA LEAKAGE!")
    
    # Remove high cardinality
    categorical_features = [col for col in categorical_features 
                           if col not in high_cardinality]
    
    print(f"\n✅ Selected Features:")
    print(f"  - Numerical: {len(numerical_features)} features")
    print(f"    {numerical_features}")
    print(f"  - Categorical: {len(categorical_features)} features")
    print(f"    {categorical_features}")
    
    print(f"\n🚫 Excluded:")
    print(f"  - Data leakage: {leakage_features}")
    print(f"  - High cardinality: {high_cardinality}")
    
    return numerical_features, categorical_features

# ============================================================================
# ANALYZE CORRELATIONS (Without efficiency_score)
# ============================================================================

def analyze_correlations_corrected(df, numerical_cols, target_col='target_high_efficiency'):
    """Correlation analysis WITHOUT the leakage feature"""
    
    print("\n" + "="*80)
    print("CORRELATION ANALYSIS (Corrected)")
    print("="*80)
    
    # Remove efficiency_score if present
    numerical_cols_clean = [col for col in numerical_cols if col != 'efficiency_score']
    
    corr_matrix = df[numerical_cols_clean].corr()
    target_corr = corr_matrix[target_col].drop(target_col).sort_values(ascending=False)
    
    print(f"\n🎯 Feature Correlation with Target (WITHOUT efficiency_score):")
    print(target_corr)
    
    # Identify important features
    important_features = target_corr[abs(target_corr) > 0.05].index.tolist()
    if important_features:
        print(f"\n✅ Important features (|corr| > 0.05):")
        for feat in important_features:
            print(f"  - {feat}: {target_corr[feat]:.4f}")
    else:
        print(f"\n⚠️  No features with |correlation| > 0.05")
        print(f"   Top 5 features by correlation:")
        for feat in target_corr.head(5).index:
            print(f"  - {feat}: {target_corr[feat]:.4f}")
    
    # Plot
    plt.figure(figsize=(10, 8))
    top_15 = target_corr.head(15) if len(target_corr) > 15 else target_corr
    colors = ['green' if x > 0 else 'red' for x in top_15]
    plt.barh(range(len(top_15)), top_15, color=colors, alpha=0.7)
    plt.yticks(range(len(top_15)), top_15.index)
    plt.xlabel('Correlation with Target', fontsize=12)
    plt.title('Feature Correlation (Corrected - No Leakage)', fontsize=14, fontweight='bold')
    plt.axvline(x=0, color='black', linestyle='--', linewidth=0.8)
    plt.tight_layout()
    plt.savefig('target_correlation_corrected.png', dpi=300, bbox_inches='tight')
    print(f"\n📊 Saved: target_correlation_corrected.png")
    plt.close()
    
    return target_corr

# ============================================================================
# PREPROCESSING
# ============================================================================

def preprocess_data_corrected(df, numerical_features, categorical_features, 
                              target_col='target_high_efficiency', test_size=0.2, random_state=42):
    """Preprocess data for modeling"""
    
    print("\n" + "="*80)
    print("DATA PREPROCESSING")
    print("="*80)
    
    X = df[numerical_features + categorical_features].copy()
    y = df[target_col].copy()
    
    # Handle missing values
    for col in numerical_features:
        if X[col].isnull().sum() > 0:
            X[col].fillna(X[col].median(), inplace=True)
    
    for col in categorical_features:
        if X[col].isnull().sum() > 0:
            X[col].fillna(X[col].mode()[0], inplace=True)
    
    # Encode categoricals
    label_encoders = {}
    for col in categorical_features:
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col].astype(str))
        label_encoders[col] = le
    
    # Split
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )
    
    print(f"\n📊 Data Split:")
    print(f"  - Training: {X_train.shape}")
    print(f"  - Test: {X_test.shape}")
    print(f"  - Class balance: {y_train.value_counts(normalize=True).to_dict()}")
    
    # Scale numerical features
    scaler = StandardScaler()
    X_train[numerical_features] = scaler.fit_transform(X_train[numerical_features])
    X_test[numerical_features] = scaler.transform(X_test[numerical_features])
    
    print(f"✅ Scaled numerical features")
    
    return X_train, X_test, y_train, y_test, scaler, label_encoders

# ============================================================================
# TRAIN MODELS (Logistic Regression + Random Forest)
# ============================================================================

def train_models_corrected(X_train, X_test, y_train, y_test):
    """Train both baseline and improved models"""
    
    print("\n" + "="*80)
    print("MODEL TRAINING (Corrected - No Data Leakage)")
    print("="*80)
    
    results = {}
    
    # ========== MODEL 1: Logistic Regression ==========
    print(f"\n{'='*60}")
    print(f"MODEL 1: LOGISTIC REGRESSION (Baseline)")
    print(f"{'='*60}")
    
    lr_model = LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced')
    lr_model.fit(X_train, y_train)
    
    lr_pred_train = lr_model.predict(X_train)
    lr_pred_test = lr_model.predict(X_test)
    lr_pred_proba = lr_model.predict_proba(X_test)[:, 1]
    
    print(f"\n📊 Logistic Regression Performance:")
    print(f"  Training Accuracy: {accuracy_score(y_train, lr_pred_train):.4f}")
    print(f"  Test Accuracy: {accuracy_score(y_test, lr_pred_test):.4f}")
    print(f"  ROC-AUC: {roc_auc_score(y_test, lr_pred_proba):.4f}")
    
    # Cross-validation
    cv_scores = cross_val_score(lr_model, X_train, y_train, cv=5, scoring='roc_auc')
    print(f"  CV ROC-AUC (5-fold): {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})")
    
    print(f"\n📋 Classification Report:")
    print(classification_report(y_test, lr_pred_test))
    
    results['logistic_regression'] = {
        'model': lr_model,
        'accuracy': accuracy_score(y_test, lr_pred_test),
        'roc_auc': roc_auc_score(y_test, lr_pred_proba),
        'predictions': lr_pred_test,
        'probabilities': lr_pred_proba
    }
    
    # Feature importance
    lr_importance = pd.DataFrame({
        'feature': X_train.columns,
        'coefficient': lr_model.coef_[0]
    }).sort_values('coefficient', key=abs, ascending=False)
    
    print(f"\n🎯 Top 10 Features (Logistic Regression):")
    print(lr_importance.head(10))
    
    # ========== MODEL 2: Random Forest ==========
    print(f"\n{'='*60}")
    print(f"MODEL 2: RANDOM FOREST")
    print(f"{'='*60}")
    
    rf_model = RandomForestClassifier(
        n_estimators=100, 
        max_depth=10,
        min_samples_split=100,
        min_samples_leaf=50,
        random_state=42,
        class_weight='balanced',
        n_jobs=-1
    )
    rf_model.fit(X_train, y_train)
    
    rf_pred_train = rf_model.predict(X_train)
    rf_pred_test = rf_model.predict(X_test)
    rf_pred_proba = rf_model.predict_proba(X_test)[:, 1]
    
    print(f"\n📊 Random Forest Performance:")
    print(f"  Training Accuracy: {accuracy_score(y_train, rf_pred_train):.4f}")
    print(f"  Test Accuracy: {accuracy_score(y_test, rf_pred_test):.4f}")
    print(f"  ROC-AUC: {roc_auc_score(y_test, rf_pred_proba):.4f}")
    
    # Cross-validation
    cv_scores_rf = cross_val_score(rf_model, X_train, y_train, cv=5, scoring='roc_auc')
    print(f"  CV ROC-AUC (5-fold): {cv_scores_rf.mean():.4f} (+/- {cv_scores_rf.std():.4f})")
    
    print(f"\n📋 Classification Report:")
    print(classification_report(y_test, rf_pred_test))
    
    results['random_forest'] = {
        'model': rf_model,
        'accuracy': accuracy_score(y_test, rf_pred_test),
        'roc_auc': roc_auc_score(y_test, rf_pred_proba),
        'predictions': rf_pred_test,
        'probabilities': rf_pred_proba
    }
    
    # Feature importance
    rf_importance = pd.DataFrame({
        'feature': X_train.columns,
        'importance': rf_model.feature_importances_
    }).sort_values('importance', ascending=False)
    
    print(f"\n🎯 Top 10 Features (Random Forest):")
    print(rf_importance.head(10))
    
    # ========== VISUALIZATIONS ==========
    
    # Confusion matrices
    fig, axes = plt.subplots(1, 2, figsize=(14, 5))
    
    for idx, (name, result) in enumerate([('Logistic Regression', results['logistic_regression']),
                                          ('Random Forest', results['random_forest'])]):
        cm = confusion_matrix(y_test, result['predictions'])
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[idx],
                   xticklabels=['Low', 'High'], yticklabels=['Low', 'High'])
        axes[idx].set_title(f'{name}\nAccuracy: {result["accuracy"]:.4f}', fontweight='bold')
        axes[idx].set_ylabel('Actual')
        axes[idx].set_xlabel('Predicted')
    
    plt.tight_layout()
    plt.savefig('confusion_matrices_corrected.png', dpi=300, bbox_inches='tight')
    print(f"\n📊 Saved: confusion_matrices_corrected.png")
    plt.close()
    
    # ROC curves
    plt.figure(figsize=(10, 6))
    
    for name, result in results.items():
        fpr, tpr, _ = roc_curve(y_test, result['probabilities'])
        plt.plot(fpr, tpr, linewidth=2, 
                label=f'{name.replace("_", " ").title()} (AUC={result["roc_auc"]:.3f})')
    
    plt.plot([0, 1], [0, 1], 'k--', linewidth=1, label='Random')
    plt.xlabel('False Positive Rate', fontsize=12)
    plt.ylabel('True Positive Rate', fontsize=12)
    plt.title('ROC Curves (Corrected Models)', fontsize=14, fontweight='bold')
    plt.legend(fontsize=10)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig('roc_curves_corrected.png', dpi=300, bbox_inches='tight')
    print(f"📊 Saved: roc_curves_corrected.png")
    plt.close()
    
    # Feature importance comparison
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    
    # Logistic Regression
    top_lr = lr_importance.head(10)
    colors_lr = ['green' if x > 0 else 'red' for x in top_lr['coefficient']]
    axes[0].barh(range(len(top_lr)), top_lr['coefficient'], color=colors_lr, alpha=0.7)
    axes[0].set_yticks(range(len(top_lr)))
    axes[0].set_yticklabels(top_lr['feature'])
    axes[0].set_xlabel('Coefficient')
    axes[0].set_title('Logistic Regression Feature Importance', fontweight='bold')
    axes[0].axvline(x=0, color='black', linestyle='--', linewidth=0.8)
    
    # Random Forest
    top_rf = rf_importance.head(10)
    axes[1].barh(range(len(top_rf)), top_rf['importance'], color='steelblue', alpha=0.7)
    axes[1].set_yticks(range(len(top_rf)))
    axes[1].set_yticklabels(top_rf['feature'])
    axes[1].set_xlabel('Importance')
    axes[1].set_title('Random Forest Feature Importance', fontweight='bold')
    
    plt.tight_layout()
    plt.savefig('feature_importance_corrected.png', dpi=300, bbox_inches='tight')
    print(f"📊 Saved: feature_importance_corrected.png")
    plt.close()
    
    return results, lr_importance, rf_importance

# ============================================================================
# MAIN EXECUTION
# ============================================================================

def main(filepath):
    """Main execution pipeline - CORRECTED VERSION"""
    
    print("\n" + "="*80)
    print("🚗 EV EFFICIENCY PREDICTION - CORRECTED PIPELINE")
    print("   (No Data Leakage - Using Engineered Features)")
    print("="*80)
    
    # Load data
    print(f"\nLoading data from: {filepath}")
    df = pd.read_csv(filepath)
    print(f"Dataset shape: {df.shape}")
    
    # Feature engineering
    df_eng = engineer_features_corrected(df)
    
    # Feature selection (removes efficiency_score)
    numerical_features, categorical_features = select_features_corrected(df_eng)
    
    # Correlation analysis
    target_corr = analyze_correlations_corrected(df_eng, numerical_features + [df_eng.columns[df_eng.columns == 'target_high_efficiency'][0]])
    
    # Preprocessing
    X_train, X_test, y_train, y_test, scaler, label_encoders = preprocess_data_corrected(
        df_eng, numerical_features, categorical_features
    )
    
    # Train models
    results, lr_importance, rf_importance = train_models_corrected(
        X_train, X_test, y_train, y_test
    )
    
    # Summary
    print("\n" + "="*80)
    print("✅ PIPELINE COMPLETED!")
    print("="*80)
    
    print(f"\n📊 Model Comparison:")
    print(f"{'Model':<25} {'Accuracy':<12} {'ROC-AUC':<12}")
    print(f"{'-'*50}")
    for name, result in results.items():
        print(f"{name.replace('_', ' ').title():<25} {result['accuracy']:<12.4f} {result['roc_auc']:<12.4f}")
    
    print(f"\n📁 Generated Files:")
    print(f"  - target_correlation_corrected.png")
    print(f"  - confusion_matrices_corrected.png")
    print(f"  - roc_curves_corrected.png")
    print(f"  - feature_importance_corrected.png")
    
    print(f"\n💡 Next Steps:")
    if results['logistic_regression']['roc_auc'] < 0.70:
        print(f"  ⚠️  Performance is lower than expected (ROC-AUC < 0.70)")
        print(f"  → This is NORMAL after removing data leakage!")
        print(f"  → Your target might be genuinely hard to predict from these features")
        print(f"  → Consider:")
        print(f"     - Collect more relevant features (vehicle weight, motor power, etc.)")
        print(f"     - Try XGBoost/LightGBM for better performance")
        print(f"     - Check if target definition makes sense")
    else:
        print(f"  ✅ Good performance! Now:")
        print(f"     1. Try XGBoost/LightGBM")
        print(f"     2. Hyperparameter tuning")
        print(f"     3. Ensemble methods")
    
    return df_eng, results, lr_importance, rf_importance

if __name__ == "__main__":
    FILEPATH = "dataset/cleaned_dataset.csv"  # <-- CHANGE THIS
    
    df_eng, results, lr_imp, rf_imp = main(FILEPATH)
```

```python
"""
Advanced Models: XGBoost & LightGBM
====================================
This script trains gradient boosting models to improve upon the baseline.
Includes: Feature selection, hyperparameter tuning, model comparison
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import (
    classification_report, 
    confusion_matrix, 
    roc_auc_score, 
    roc_curve,
    accuracy_score,
    precision_recall_curve,
    auc
)
import xgboost as xgb
import lightgbm as lgb
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
import warnings
warnings.filterwarnings('ignore')

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

# ============================================================================
# FEATURE ENGINEERING
# ============================================================================

def engineer_features(df):
    """Create derived features"""
    
    print("\n" + "="*80)
    print("FEATURE ENGINEERING")
    print("="*80)
    
    df_eng = df.copy()
    
    # Core efficiency metrics
    df_eng['range_efficiency'] = df_eng['range_km'] / (df_eng['battery_kwh'] + 0.001)
    df_eng['battery_range_ratio'] = df_eng['battery_kwh'] / (df_eng['range_km'] + 0.001)
    df_eng['charging_speed'] = df_eng['battery_kwh'] / (df_eng['charging_time_hr'] + 0.001)
    df_eng['power_indicator'] = 100 / (df_eng['acceleration_0_100_kmph'] + 0.001)
    df_eng['speed_efficiency'] = df_eng['range_km'] / (df_eng['top_speed_kmph'] + 0.001)
    df_eng['vehicle_age'] = 2026 - df_eng['release_year']
    df_eng['price_per_kwh'] = df_eng['price_usd'] / (df_eng['battery_kwh'] + 0.001)
    df_eng['range_per_seat'] = df_eng['range_km'] / (df_eng['seats'] + 0.001)
    df_eng['is_electric'] = (df_eng['fuel_type'] == 'Electric').astype(int)
    df_eng['has_fast_charging'] = df_eng['fast_charging'].astype(int)
    
    print("✅ Created 10 engineered features")
    
    return df_eng

# ============================================================================
# FEATURE SELECTION (Multiple Strategies)
# ============================================================================

def select_features(df, target_col='target_high_efficiency', strategy='optimal'):
    """
    Select features based on different strategies
    
    Strategies:
    - 'all': All features except leakage (26 features)
    - 'optimal': Features with |corr| > 0.1 (8-10 features)
    - 'minimal': Top 5 features only (5 features)
    """
    
    print("\n" + "="*80)
    print(f"FEATURE SELECTION - Strategy: {strategy.upper()}")
    print("="*80)
    
    # Always exclude these
    exclude = ['manufacturer', 'model', 'city', 'color', 'efficiency_score']
    
    if strategy == 'minimal':
        # Top 5 most important features
        numerical_features = [
            'range_efficiency',
            'range_km',
            'speed_efficiency',
        ]
        categorical_features = ['type', 'fuel_type']
        
    elif strategy == 'optimal':
        # Features with |correlation| > 0.1
        numerical_features = [
            'range_efficiency',
            'range_km', 
            'speed_efficiency',
            'range_per_seat',
            'battery_kwh',  # For interactions
            'price_usd',     # Weak but might help
        ]
        categorical_features = ['type', 'drive_type', 'fuel_type', 'country']
        
    else:  # 'all'
        # All numerical features except exclusions
        numerical_features = [col for col in df.select_dtypes(include=[np.number]).columns 
                            if col not in exclude + [target_col]]
        categorical_features = [col for col in df.select_dtypes(include=['object']).columns 
                              if col not in exclude]
    
    print(f"\n📊 Selected Features:")
    print(f"  - Numerical: {len(numerical_features)} features")
    print(f"    {numerical_features}")
    print(f"  - Categorical: {len(categorical_features)} features")
    print(f"    {categorical_features}")
    print(f"  - Total: {len(numerical_features) + len(categorical_features)} features")
    
    return numerical_features, categorical_features

# ============================================================================
# DATA PREPROCESSING
# ============================================================================

def preprocess_data(df, numerical_features, categorical_features, 
                   target_col='target_high_efficiency', test_size=0.2, random_state=42):
    """Preprocess data for modeling"""
    
    print("\n" + "="*80)
    print("DATA PREPROCESSING")
    print("="*80)
    
    X = df[numerical_features + categorical_features].copy()
    y = df[target_col].copy()
    
    # Handle missing values
    for col in numerical_features:
        if X[col].isnull().sum() > 0:
            X[col].fillna(X[col].median(), inplace=True)
    
    for col in categorical_features:
        if X[col].isnull().sum() > 0:
            X[col].fillna(X[col].mode()[0], inplace=True)
    
    # Encode categorical variables
    label_encoders = {}
    for col in categorical_features:
        le = LabelEncoder()
        X[col] = le.fit_transform(X[col].astype(str))
        label_encoders[col] = le
    
    # Train-test split
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=test_size, random_state=random_state, stratify=y
    )
    
    print(f"\n📊 Data Split:")
    print(f"  - Training: {X_train.shape}")
    print(f"  - Test: {X_test.shape}")
    print(f"  - Class balance: {y_train.value_counts(normalize=True).to_dict()}")
    
    # Scale numerical features (for Logistic Regression comparison)
    scaler = StandardScaler()
    X_train_scaled = X_train.copy()
    X_test_scaled = X_test.copy()
    X_train_scaled[numerical_features] = scaler.fit_transform(X_train[numerical_features])
    X_test_scaled[numerical_features] = scaler.transform(X_test[numerical_features])
    
    print(f"✅ Preprocessing completed")
    
    return X_train, X_test, y_train, y_test, X_train_scaled, X_test_scaled, scaler, label_encoders

# ============================================================================
# TRAIN ALL MODELS
# ============================================================================

def train_all_models(X_train, X_test, y_train, y_test, X_train_scaled, X_test_scaled):
    """Train Logistic Regression, Random Forest, XGBoost, and LightGBM"""
    
    print("\n" + "="*80)
    print("MODEL TRAINING")
    print("="*80)
    
    results = {}
    
    # ========== MODEL 1: Logistic Regression (Baseline) ==========
    print(f"\n{'='*60}")
    print(f"MODEL 1: LOGISTIC REGRESSION (Baseline)")
    print(f"{'='*60}")
    
    lr_model = LogisticRegression(max_iter=1000, random_state=42, class_weight='balanced')
    lr_model.fit(X_train_scaled, y_train)
    
    lr_pred_test = lr_model.predict(X_test_scaled)
    lr_pred_proba = lr_model.predict_proba(X_test_scaled)[:, 1]
    
    results['Logistic Regression'] = {
        'model': lr_model,
        'accuracy': accuracy_score(y_test, lr_pred_test),
        'roc_auc': roc_auc_score(y_test, lr_pred_proba),
        'predictions': lr_pred_test,
        'probabilities': lr_pred_proba
    }
    
    print(f"  Accuracy: {results['Logistic Regression']['accuracy']:.4f}")
    print(f"  ROC-AUC:  {results['Logistic Regression']['roc_auc']:.4f}")
    
    # ========== MODEL 2: Random Forest ==========
    print(f"\n{'='*60}")
    print(f"MODEL 2: RANDOM FOREST")
    print(f"{'='*60}")
    
    rf_model = RandomForestClassifier(
        n_estimators=100,
        max_depth=10,
        min_samples_split=100,
        min_samples_leaf=50,
        random_state=42,
        class_weight='balanced',
        n_jobs=-1
    )
    rf_model.fit(X_train, y_train)
    
    rf_pred_test = rf_model.predict(X_test)
    rf_pred_proba = rf_model.predict_proba(X_test)[:, 1]
    
    results['Random Forest'] = {
        'model': rf_model,
        'accuracy': accuracy_score(y_test, rf_pred_test),
        'roc_auc': roc_auc_score(y_test, rf_pred_proba),
        'predictions': rf_pred_test,
        'probabilities': rf_pred_proba
    }
    
    print(f"  Accuracy: {results['Random Forest']['accuracy']:.4f}")
    print(f"  ROC-AUC:  {results['Random Forest']['roc_auc']:.4f}")
    
    # ========== MODEL 3: XGBoost ==========
    print(f"\n{'='*60}")
    print(f"MODEL 3: XGBOOST")
    print(f"{'='*60}")
    
    xgb_model = xgb.XGBClassifier(
        n_estimators=200,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        eval_metric='logloss',
        use_label_encoder=False
    )
    xgb_model.fit(X_train, y_train)
    
    xgb_pred_test = xgb_model.predict(X_test)
    xgb_pred_proba = xgb_model.predict_proba(X_test)[:, 1]
    
    results['XGBoost'] = {
        'model': xgb_model,
        'accuracy': accuracy_score(y_test, xgb_pred_test),
        'roc_auc': roc_auc_score(y_test, xgb_pred_proba),
        'predictions': xgb_pred_test,
        'probabilities': xgb_pred_proba
    }
    
    print(f"  Accuracy: {results['XGBoost']['accuracy']:.4f}")
    print(f"  ROC-AUC:  {results['XGBoost']['roc_auc']:.4f}")
    
    # ========== MODEL 4: LightGBM ==========
    print(f"\n{'='*60}")
    print(f"MODEL 4: LIGHTGBM")
    print(f"{'='*60}")
    
    lgb_model = lgb.LGBMClassifier(
        n_estimators=200,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        random_state=42,
        verbose=-1
    )
    lgb_model.fit(X_train, y_train)
    
    lgb_pred_test = lgb_model.predict(X_test)
    lgb_pred_proba = lgb_model.predict_proba(X_test)[:, 1]
    
    results['LightGBM'] = {
        'model': lgb_model,
        'accuracy': accuracy_score(y_test, lgb_pred_test),
        'roc_auc': roc_auc_score(y_test, lgb_pred_proba),
        'predictions': lgb_pred_test,
        'probabilities': lgb_pred_proba
    }
    
    print(f"  Accuracy: {results['LightGBM']['accuracy']:.4f}")
    print(f"  ROC-AUC:  {results['LightGBM']['roc_auc']:.4f}")
    
    return results

# ============================================================================
# HYPERPARAMETER TUNING (XGBoost & LightGBM)
# ============================================================================

def tune_models(X_train, y_train, model_type='xgboost'):
    """
    Hyperparameter tuning using GridSearchCV
    
    model_type: 'xgboost' or 'lightgbm'
    """
    
    print(f"\n{'='*60}")
    print(f"HYPERPARAMETER TUNING - {model_type.upper()}")
    print(f"{'='*60}")
    print("⏳ This may take a few minutes...")
    
    if model_type == 'xgboost':
        base_model = xgb.XGBClassifier(
            random_state=42,
            eval_metric='logloss',
            use_label_encoder=False
        )
        param_grid = {
            'n_estimators': [100, 200, 300],
            'max_depth': [4, 6, 8],
            'learning_rate': [0.01, 0.05, 0.1],
            'subsample': [0.8],
            'colsample_bytree': [0.8]
        }
    else:  # lightgbm
        base_model = lgb.LGBMClassifier(
            random_state=42,
            verbose=-1
        )
        param_grid = {
            'n_estimators': [100, 200, 300],
            'max_depth': [4, 6, 8],
            'learning_rate': [0.01, 0.05, 0.1],
            'subsample': [0.8],
            'colsample_bytree': [0.8]
        }
    
    grid_search = GridSearchCV(
        base_model,
        param_grid,
        cv=3,  # 3-fold CV for speed
        scoring='roc_auc',
        n_jobs=-1,
        verbose=0
    )
    
    grid_search.fit(X_train, y_train)
    
    print(f"\n✅ Best Parameters:")
    for param, value in grid_search.best_params_.items():
        print(f"  - {param}: {value}")
    print(f"\n✅ Best CV ROC-AUC: {grid_search.best_score_:.4f}")
    
    return grid_search.best_estimator_, grid_search.best_params_

# ============================================================================
# VISUALIZATIONS
# ============================================================================

def create_visualizations(results, y_test, X_train, feature_names):
    """Create comprehensive visualizations"""
    
    print("\n" + "="*80)
    print("GENERATING VISUALIZATIONS")
    print("="*80)
    
    # 1. Model Comparison Bar Chart
    plt.figure(figsize=(12, 6))
    models = list(results.keys())
    accuracies = [results[m]['accuracy'] for m in models]
    roc_aucs = [results[m]['roc_auc'] for m in models]
    
    x = np.arange(len(models))
    width = 0.35
    
    plt.bar(x - width/2, accuracies, width, label='Accuracy', alpha=0.8)
    plt.bar(x + width/2, roc_aucs, width, label='ROC-AUC', alpha=0.8)
    
    plt.xlabel('Models', fontsize=12)
    plt.ylabel('Score', fontsize=12)
    plt.title('Model Performance Comparison', fontsize=14, fontweight='bold')
    plt.xticks(x, models, rotation=15, ha='right')
    plt.ylim(0.75, 0.95)
    plt.legend()
    plt.grid(axis='y', alpha=0.3)
    plt.tight_layout()
    plt.savefig('model_comparison.png', dpi=300, bbox_inches='tight')
    print("📊 Saved: model_comparison.png")
    plt.close()
    
    # 2. ROC Curves (All Models)
    plt.figure(figsize=(10, 8))
    
    for name, result in results.items():
        fpr, tpr, _ = roc_curve(y_test, result['probabilities'])
        plt.plot(fpr, tpr, linewidth=2, label=f'{name} (AUC={result["roc_auc"]:.3f})')
    
    plt.plot([0, 1], [0, 1], 'k--', linewidth=1, label='Random')
    plt.xlabel('False Positive Rate', fontsize=12)
    plt.ylabel('True Positive Rate', fontsize=12)
    plt.title('ROC Curves - All Models', fontsize=14, fontweight='bold')
    plt.legend(fontsize=10, loc='lower right')
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.savefig('roc_curves_all_models.png', dpi=300, bbox_inches='tight')
    print("📊 Saved: roc_curves_all_models.png")
    plt.close()
    
    # 3. Confusion Matrices (2x2 Grid)
    fig, axes = plt.subplots(2, 2, figsize=(14, 12))
    axes = axes.ravel()
    
    for idx, (name, result) in enumerate(results.items()):
        cm = confusion_matrix(y_test, result['predictions'])
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', ax=axes[idx],
                   xticklabels=['Low', 'High'], yticklabels=['Low', 'High'])
        axes[idx].set_title(f'{name}\nAcc: {result["accuracy"]:.4f} | AUC: {result["roc_auc"]:.4f}', 
                           fontweight='bold')
        axes[idx].set_ylabel('Actual')
        axes[idx].set_xlabel('Predicted')
    
    plt.tight_layout()
    plt.savefig('confusion_matrices_all_models.png', dpi=300, bbox_inches='tight')
    print("📊 Saved: confusion_matrices_all_models.png")
    plt.close()
    
    # 4. Feature Importance (XGBoost and LightGBM)
    fig, axes = plt.subplots(1, 2, figsize=(16, 6))
    
    # XGBoost
    if 'XGBoost' in results:
        xgb_importance = pd.DataFrame({
            'feature': feature_names,
            'importance': results['XGBoost']['model'].feature_importances_
        }).sort_values('importance', ascending=False).head(15)
        
        axes[0].barh(range(len(xgb_importance)), xgb_importance['importance'], 
                    color='steelblue', alpha=0.7)
        axes[0].set_yticks(range(len(xgb_importance)))
        axes[0].set_yticklabels(xgb_importance['feature'])
        axes[0].set_xlabel('Importance', fontsize=12)
        axes[0].set_title('XGBoost Feature Importance (Top 15)', fontsize=12, fontweight='bold')
        axes[0].invert_yaxis()
    
    # LightGBM
    if 'LightGBM' in results:
        lgb_importance = pd.DataFrame({
            'feature': feature_names,
            'importance': results['LightGBM']['model'].feature_importances_
        }).sort_values('importance', ascending=False).head(15)
        
        axes[1].barh(range(len(lgb_importance)), lgb_importance['importance'], 
                    color='darkorange', alpha=0.7)
        axes[1].set_yticks(range(len(lgb_importance)))
        axes[1].set_yticklabels(lgb_importance['feature'])
        axes[1].set_xlabel('Importance', fontsize=12)
        axes[1].set_title('LightGBM Feature Importance (Top 15)', fontsize=12, fontweight='bold')
        axes[1].invert_yaxis()
    
    plt.tight_layout()
    plt.savefig('feature_importance_gbm.png', dpi=300, bbox_inches='tight')
    print("📊 Saved: feature_importance_gbm.png")
    plt.close()
    
    print("✅ All visualizations saved!")

# ============================================================================
# MAIN EXECUTION
# ============================================================================

def main(filepath, feature_strategy='optimal', tune_hyperparams=False):
    """
    Main execution pipeline
    
    Args:
        filepath: Path to dataset
        feature_strategy: 'minimal', 'optimal', or 'all'
        tune_hyperparams: Whether to perform hyperparameter tuning (slow!)
    """
    
    print("\n" + "="*80)
    print("🚀 ADVANCED MODELS: XGBoost & LightGBM")
    print("="*80)
    
    # Load data
    print(f"\nLoading data from: {filepath}")
    df = pd.read_csv(filepath)
    print(f"Dataset shape: {df.shape}")
    
    # Feature engineering
    df_eng = engineer_features(df)
    
    # Feature selection
    numerical_features, categorical_features = select_features(
        df_eng, strategy=feature_strategy
    )
    
    # Preprocessing
    X_train, X_test, y_train, y_test, X_train_scaled, X_test_scaled, scaler, label_encoders = \
        preprocess_data(df_eng, numerical_features, categorical_features)
    
    # Train all models
    results = train_all_models(
        X_train, X_test, y_train, y_test, X_train_scaled, X_test_scaled
    )
    
    # Optional: Hyperparameter tuning
    if tune_hyperparams:
        print("\n" + "="*80)
        print("HYPERPARAMETER TUNING")
        print("="*80)
        
        # Tune XGBoost
        best_xgb, best_xgb_params = tune_models(X_train, y_train, 'xgboost')
        xgb_tuned_pred = best_xgb.predict(X_test)
        xgb_tuned_proba = best_xgb.predict_proba(X_test)[:, 1]
        
        results['XGBoost (Tuned)'] = {
            'model': best_xgb,
            'accuracy': accuracy_score(y_test, xgb_tuned_pred),
            'roc_auc': roc_auc_score(y_test, xgb_tuned_proba),
            'predictions': xgb_tuned_pred,
            'probabilities': xgb_tuned_proba
        }
        
        # Tune LightGBM
        best_lgb, best_lgb_params = tune_models(X_train, y_train, 'lightgbm')
        lgb_tuned_pred = best_lgb.predict(X_test)
        lgb_tuned_proba = best_lgb.predict_proba(X_test)[:, 1]
        
        results['LightGBM (Tuned)'] = {
            'model': best_lgb,
            'accuracy': accuracy_score(y_test, lgb_tuned_pred),
            'roc_auc': roc_auc_score(y_test, lgb_tuned_proba),
            'predictions': lgb_tuned_pred,
            'probabilities': lgb_tuned_proba
        }
    
    # Visualizations
    feature_names = numerical_features + categorical_features
    create_visualizations(results, y_test, X_train, feature_names)
    
    # Final Summary
    print("\n" + "="*80)
    print("✅ PIPELINE COMPLETED!")
    print("="*80)
    
    print(f"\n📊 FINAL MODEL COMPARISON:")
    print(f"{'Model':<25} {'Accuracy':<12} {'ROC-AUC':<12} {'Improvement'}")
    print(f"{'-'*65}")
    
    baseline_auc = results['Logistic Regression']['roc_auc']
    for name, result in sorted(results.items(), key=lambda x: x[1]['roc_auc'], reverse=True):
        improvement = result['roc_auc'] - baseline_auc
        print(f"{name:<25} {result['accuracy']:<12.4f} {result['roc_auc']:<12.4f} "
              f"{'+' if improvement >= 0 else ''}{improvement:.4f}")
    
    # Best model
    best_model_name = max(results, key=lambda x: results[x]['roc_auc'])
    best_model = results[best_model_name]['model']
    
    print(f"\n🏆 BEST MODEL: {best_model_name}")
    print(f"   ROC-AUC: {results[best_model_name]['roc_auc']:.4f}")
    print(f"   Accuracy: {results[best_model_name]['accuracy']:.4f}")
    
    # Feature importance for best model
    if hasattr(best_model, 'feature_importances_'):
        print(f"\n🎯 Top 10 Features ({best_model_name}):")
        importance_df = pd.DataFrame({
            'feature': feature_names,
            'importance': best_model.feature_importances_
        }).sort_values('importance', ascending=False)
        print(importance_df.head(10).to_string(index=False))
    
    print(f"\n📁 Generated Files:")
    print(f"  - model_comparison.png")
    print(f"  - roc_curves_all_models.png")
    print(f"  - confusion_matrices_all_models.png")
    print(f"  - feature_importance_gbm.png")
    
    print(f"\n💡 Next Steps:")
    print(f"  1. Review feature importance")
    print(f"  2. Try different feature strategies ('minimal', 'optimal', 'all')")
    print(f"  3. Enable hyperparameter tuning (set tune_hyperparams=True)")
    print(f"  4. Save best model for production")
    print(f"  5. Create ensemble (combine top models)")
    
    return df_eng, results, best_model

# ============================================================================
# USAGE
# ============================================================================

if __name__ == "__main__":
    
    # Configuration
    FILEPATH = "dataset/cleaned_dataset.csv"  # <-- UPDATE THIS
    FEATURE_STRATEGY = 'optimal'  # Options: 'minimal', 'optimal', 'all'
    TUNE_HYPERPARAMS = False  # Set True for hyperparameter tuning (slower)
    
    # Run pipeline
    df_engineered, results, best_model = main(
        filepath=FILEPATH,
        feature_strategy=FEATURE_STRATEGY,
        tune_hyperparams=TUNE_HYPERPARAMS
    )
    
    print("\n" + "="*80)
    print("🎉 ALL DONE!")
    print("="*80)
```

```python
"""
Production-Ready End-to-End ML Pipeline
========================================
Best practices implemented:
1. Train/Val/Test split BEFORE any preprocessing
2. No data leakage from scaling/encoding
3. Proper validation for model selection
4. Complete experiment tracking
5. Model comparison with statistical tests
6. Ready for team collaboration and deployment
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, cross_val_score, StratifiedKFold
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    classification_report, 
    confusion_matrix, 
    roc_auc_score, 
    roc_curve,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
import xgboost as xgb
import lightgbm as lgb
import pickle
import json
from datetime import datetime
import warnings
warnings.filterwarnings('ignore')

sns.set_style("whitegrid")
plt.rcParams['figure.figsize'] = (12, 8)

# ============================================================================
# CONFIGURATION
# ============================================================================

class Config:
    """Central configuration for the entire pipeline"""
    
    # Data paths
    DATA_PATH = "dataset/cleaned_dataset.csv"
    OUTPUT_DIR = "outputs"
    MODEL_DIR = "models"
    
    # Data splits (BEFORE any preprocessing!)
    TRAIN_SIZE = 0.70  # 70% training
    VAL_SIZE = 0.15    # 15% validation
    TEST_SIZE = 0.15   # 15% test
    RANDOM_STATE = 42
    
    # Feature selection strategy
    FEATURE_STRATEGY = 'optimal'  # 'minimal', 'optimal', or 'all'
    
    # Data leakage prevention
    EXCLUDE_FEATURES = ['efficiency_score', 'manufacturer', 'model', 'city', 'color']
    
    # Model hyperparameters
    MODELS_CONFIG = {
        'logistic_regression': {
            'max_iter': 1000,
            'random_state': RANDOM_STATE,
            'class_weight': 'balanced'
        },
        'random_forest': {
            'n_estimators': 100,
            'max_depth': 10,
            'min_samples_split': 100,
            'min_samples_leaf': 50,
            'random_state': RANDOM_STATE,
            'class_weight': 'balanced',
            'n_jobs': -1
        },
        'xgboost': {
            'n_estimators': 200,
            'max_depth': 6,
            'learning_rate': 0.05,
            'subsample': 0.8,
            'colsample_bytree': 0.8,
            'random_state': RANDOM_STATE,
            'eval_metric': 'logloss',
            'use_label_encoder': False
        },
        'lightgbm': {
            'n_estimators': 200,
            'max_depth': 6,
            'learning_rate': 0.05,
            'subsample': 0.8,
            'colsample_bytree': 0.8,
            'random_state': RANDOM_STATE,
            'verbose': -1
        }
    }
    
    # Cross-validation
    CV_FOLDS = 5
    
    # Experiment tracking
    EXPERIMENT_NAME = f"ev_efficiency_{datetime.now().strftime('%Y%m%d_%H%M%S')}"

# ============================================================================
# STEP 1: DATA SPLITTING (BEFORE ANY PREPROCESSING!)
# ============================================================================

def split_data_before_preprocessing(df, target_col='target_high_efficiency'):
    """
    Split data into train/val/test BEFORE any preprocessing
    This prevents data leakage from scaling, encoding, etc.
    """
    
    print("\n" + "="*80)
    print("STEP 1: DATA SPLITTING (Before Preprocessing)")
    print("="*80)
    
    # First split: train+val vs test
    train_val_df, test_df = train_test_split(
        df,
        test_size=Config.TEST_SIZE,
        random_state=Config.RANDOM_STATE,
        stratify=df[target_col]
    )
    
    # Second split: train vs val
    # Adjust validation size relative to train+val
    val_size_adjusted = Config.VAL_SIZE / (Config.TRAIN_SIZE + Config.VAL_SIZE)
    
    train_df, val_df = train_test_split(
        train_val_df,
        test_size=val_size_adjusted,
        random_state=Config.RANDOM_STATE,
        stratify=train_val_df[target_col]
    )
    
    print(f"\n📊 Data Split Summary:")
    print(f"  Total samples: {len(df):,}")
    print(f"  Training:   {len(train_df):,} ({len(train_df)/len(df)*100:.1f}%)")
    print(f"  Validation: {len(val_df):,} ({len(val_df)/len(df)*100:.1f}%)")
    print(f"  Test:       {len(test_df):,} ({len(test_df)/len(df)*100:.1f}%)")
    
    print(f"\n📊 Target Distribution:")
    for split_name, split_df in [('Train', train_df), ('Val', val_df), ('Test', test_df)]:
        dist = split_df[target_col].value_counts(normalize=True)
        print(f"  {split_name}: Class 0: {dist[0]:.3f}, Class 1: {dist[1]:.3f}")
    
    print(f"\n✅ Data split completed BEFORE any preprocessing (no leakage!)")
    
    return train_df, val_df, test_df

# ============================================================================
# STEP 2: FEATURE ENGINEERING
# ============================================================================

def engineer_features(df):
    """Create derived features"""
    
    df_eng = df.copy()
    
    # Core efficiency metrics
    df_eng['range_efficiency'] = df_eng['range_km'] / (df_eng['battery_kwh'] + 0.001)
    df_eng['battery_range_ratio'] = df_eng['battery_kwh'] / (df_eng['range_km'] + 0.001)
    df_eng['charging_speed'] = df_eng['battery_kwh'] / (df_eng['charging_time_hr'] + 0.001)
    df_eng['power_indicator'] = 100 / (df_eng['acceleration_0_100_kmph'] + 0.001)
    df_eng['speed_efficiency'] = df_eng['range_km'] / (df_eng['top_speed_kmph'] + 0.001)
    df_eng['vehicle_age'] = 2026 - df_eng['release_year']
    df_eng['price_per_kwh'] = df_eng['price_usd'] / (df_eng['battery_kwh'] + 0.001)
    df_eng['range_per_seat'] = df_eng['range_km'] / (df_eng['seats'] + 0.001)
    df_eng['is_electric'] = (df_eng['fuel_type'] == 'Electric').astype(int)
    df_eng['has_fast_charging'] = df_eng['fast_charging'].astype(int)
    
    return df_eng

# ============================================================================
# STEP 3: FEATURE SELECTION
# ============================================================================

def select_features(strategy='optimal'):
    """Define features based on strategy"""
    
    print("\n" + "="*80)
    print(f"STEP 2: FEATURE SELECTION - Strategy: {strategy.upper()}")
    print("="*80)
    
    if strategy == 'minimal':
        numerical_features = [
            'range_efficiency',
            'range_km',
            'speed_efficiency',
        ]
        categorical_features = ['type', 'fuel_type']
        
    elif strategy == 'optimal':
        numerical_features = [
            'range_efficiency',
            'range_km',
            'speed_efficiency',
            'range_per_seat',
            'battery_kwh',
            'price_usd',
        ]
        categorical_features = ['type', 'drive_type', 'fuel_type', 'country']
        
    else:  # 'all'
        # Will be determined after engineering
        numerical_features = None
        categorical_features = None
    
    print(f"\n✅ Selected Features:")
    if numerical_features:
        print(f"  Numerical: {numerical_features}")
        print(f"  Categorical: {categorical_features}")
        print(f"  Total: {len(numerical_features) + len(categorical_features)}")
    
    return numerical_features, categorical_features

# ============================================================================
# STEP 4: PREPROCESSING (Train on TRAIN only!)
# ============================================================================

class DataPreprocessor:
    """
    Preprocessing pipeline that learns from training data only
    Then transforms val and test sets
    """
    
    def __init__(self, numerical_features, categorical_features, target_col='target_high_efficiency'):
        self.numerical_features = numerical_features
        self.categorical_features = categorical_features
        self.target_col = target_col
        self.scaler = StandardScaler()
        self.label_encoders = {}
        self.is_fitted = False
    
    def fit(self, train_df):
        """Fit preprocessing on training data ONLY"""
        
        print("\n" + "="*80)
        print("STEP 3: PREPROCESSING (Learning from TRAIN set only)")
        print("="*80)
        
        # Handle missing values (learn from train)
        self.num_medians = {}
        self.cat_modes = {}
        
        for col in self.numerical_features:
            self.num_medians[col] = train_df[col].median()
        
        for col in self.categorical_features:
            self.cat_modes[col] = train_df[col].mode()[0]
        
        # Fit label encoders on training data
        for col in self.categorical_features:
            le = LabelEncoder()
            # Fit on train data (filling missing first)
            train_col = train_df[col].fillna(self.cat_modes[col])
            le.fit(train_col.astype(str))
            self.label_encoders[col] = le
        
        # Fit scaler on training data
        train_X = self._prepare_features(train_df)
        self.scaler.fit(train_X[self.numerical_features])
        
        self.is_fitted = True
        print(f"✅ Preprocessing fitted on training data")
        
        return self
    
    def transform(self, df, split_name=''):
        """Transform any dataset using learned preprocessing"""
        
        if not self.is_fitted:
            raise ValueError("Must call fit() before transform()")
        
        X = self._prepare_features(df)
        
        # Scale numerical features
        X[self.numerical_features] = self.scaler.transform(X[self.numerical_features])
        
        if split_name:
            print(f"✅ Transformed {split_name} set: {X.shape}")
        
        return X, df[self.target_col]
    
    def _prepare_features(self, df):
        """Prepare features (fill missing, encode)"""
        
        X = df[self.numerical_features + self.categorical_features].copy()
        
        # Fill missing values (using learned statistics)
        for col in self.numerical_features:
            X[col].fillna(self.num_medians[col], inplace=True)
        
        for col in self.categorical_features:
            X[col].fillna(self.cat_modes[col], inplace=True)
        
        # Encode categorical features (using learned encoders)
        for col in self.categorical_features:
            # Handle unseen categories
            X[col] = X[col].astype(str)
            # Replace unseen values with most common training value
            mask = ~X[col].isin(self.label_encoders[col].classes_)
            if mask.any():
                X.loc[mask, col] = self.label_encoders[col].classes_[0]
            X[col] = self.label_encoders[col].transform(X[col])
        
        return X

# ============================================================================
# STEP 5: MODEL TRAINING & EVALUATION
# ============================================================================

class ModelTrainer:
    """Train and evaluate multiple models"""
    
    def __init__(self):
        self.models = {}
        self.results = {}
    
    def train_all_models(self, X_train, y_train, X_val, y_val):
        """Train all models and evaluate on validation set"""
        
        print("\n" + "="*80)
        print("STEP 4: MODEL TRAINING & VALIDATION")
        print("="*80)
        
        model_configs = {
            'Logistic Regression': (LogisticRegression, Config.MODELS_CONFIG['logistic_regression'], True),
            'Random Forest': (RandomForestClassifier, Config.MODELS_CONFIG['random_forest'], False),
            'XGBoost': (xgb.XGBClassifier, Config.MODELS_CONFIG['xgboost'], False),
            'LightGBM': (lgb.LGBMClassifier, Config.MODELS_CONFIG['lightgbm'], False)
        }
        
        for model_name, (ModelClass, params, needs_scaling) in model_configs.items():
            print(f"\n{'='*60}")
            print(f"Training: {model_name}")
            print(f"{'='*60}")
            
            # Train model
            model = ModelClass(**params)
            
            # For logistic regression, we already scaled in preprocessing
            model.fit(X_train, y_train)
            
            # Predictions
            train_pred = model.predict(X_train)
            train_proba = model.predict_proba(X_train)[:, 1]
            
            val_pred = model.predict(X_val)
            val_proba = model.predict_proba(X_val)[:, 1]
            
            # Cross-validation on training set
            cv = StratifiedKFold(n_splits=Config.CV_FOLDS, shuffle=True, random_state=Config.RANDOM_STATE)
            cv_scores = cross_val_score(model, X_train, y_train, cv=cv, scoring='roc_auc')
            
            # Store results
            self.models[model_name] = model
            self.results[model_name] = {
                'train_accuracy': accuracy_score(y_train, train_pred),
                'train_roc_auc': roc_auc_score(y_train, train_proba),
                'val_accuracy': accuracy_score(y_val, val_pred),
                'val_roc_auc': roc_auc_score(y_val, val_proba),
                'val_precision': precision_score(y_val, val_pred),
                'val_recall': recall_score(y_val, val_pred),
                'val_f1': f1_score(y_val, val_pred),
                'cv_roc_auc_mean': cv_scores.mean(),
                'cv_roc_auc_std': cv_scores.std(),
                'val_predictions': val_pred,
                'val_probabilities': val_proba,
                'overfitting_gap': abs(roc_auc_score(y_train, train_proba) - roc_auc_score(y_val, val_proba))
            }
            
            print(f"  Train Accuracy: {self.results[model_name]['train_accuracy']:.4f}")
            print(f"  Train ROC-AUC:  {self.results[model_name]['train_roc_auc']:.4f}")
            print(f"  Val Accuracy:   {self.results[model_name]['val_accuracy']:.4f}")
            print(f"  Val ROC-AUC:    {self.results[model_name]['val_roc_auc']:.4f}")
            print(f"  CV ROC-AUC:     {self.results[model_name]['cv_roc_auc_mean']:.4f} ± {self.results[model_name]['cv_roc_auc_std']:.4f}")
            print(f"  Overfitting Gap: {self.results[model_name]['overfitting_gap']:.4f}")
        
        return self.models, self.results
    
    def evaluate_on_test(self, X_test, y_test):
        """Final evaluation on held-out test set"""
        
        print("\n" + "="*80)
        print("STEP 5: FINAL EVALUATION ON TEST SET")
        print("="*80)
        print("⚠️  TEST SET USED ONLY ONCE FOR FINAL EVALUATION")
        
        test_results = {}
        
        for model_name, model in self.models.items():
            test_pred = model.predict(X_test)
            test_proba = model.predict_proba(X_test)[:, 1]
            
            test_results[model_name] = {
                'accuracy': accuracy_score(y_test, test_pred),
                'roc_auc': roc_auc_score(y_test, test_proba),
                'precision': precision_score(y_test, test_pred),
                'recall': recall_score(y_test, test_pred),
                'f1': f1_score(y_test, test_pred),
                'predictions': test_pred,
                'probabilities': test_proba
            }
            
            print(f"\n{model_name}:")
            print(f"  Accuracy:  {test_results[model_name]['accuracy']:.4f}")
            print(f"  ROC-AUC:   {test_results[model_name]['roc_auc']:.4f}")
            print(f"  Precision: {test_results[model_name]['precision']:.4f}")
            print(f"  Recall:    {test_results[model_name]['recall']:.4f}")
            print(f"  F1-Score:  {test_results[model_name]['f1']:.4f}")
        
        return test_results

# ============================================================================
# STEP 6: MODEL COMPARISON & SELECTION
# ============================================================================

def compare_and_select_model(val_results, test_results, y_val):
    """Compare models and select the best one based on validation performance"""
    
    print("\n" + "="*80)
    print("STEP 6: MODEL COMPARISON & SELECTION")
    print("="*80)
    
    # Create comparison table
    print(f"\n📊 VALIDATION SET PERFORMANCE (for model selection):")
    print(f"{'Model':<25} {'Val ROC-AUC':<12} {'CV ROC-AUC':<15} {'Overfit Gap':<12}")
    print(f"{'-'*70}")
    
    for model_name in sorted(val_results.keys(), key=lambda x: val_results[x]['val_roc_auc'], reverse=True):
        r = val_results[model_name]
        print(f"{model_name:<25} {r['val_roc_auc']:<12.4f} "
              f"{r['cv_roc_auc_mean']:.4f}±{r['cv_roc_auc_std']:.4f}    {r['overfitting_gap']:<12.4f}")
    
    # Select best model based on validation performance
    best_model_name = max(val_results, key=lambda x: val_results[x]['val_roc_auc'])
    
    print(f"\n🏆 SELECTED MODEL (based on validation): {best_model_name}")
    print(f"   Validation ROC-AUC: {val_results[best_model_name]['val_roc_auc']:.4f}")
    
    # Show test performance for selected model
    print(f"\n📊 TEST SET PERFORMANCE (final unbiased estimate):")
    print(f"{'Model':<25} {'Test ROC-AUC':<12} {'Test Accuracy':<15}")
    print(f"{'-'*55}")
    
    for model_name in sorted(test_results.keys(), key=lambda x: test_results[x]['roc_auc'], reverse=True):
        r = test_results[model_name]
        marker = "👑" if model_name == best_model_name else "  "
        print(f"{marker} {model_name:<23} {r['roc_auc']:<12.4f} {r['accuracy']:<15.4f}")
    
    # Detailed report for best model
    print(f"\n📋 DETAILED REPORT - {best_model_name}:")
    print(f"{'='*60}")
    print(f"Validation Set:")
    print(f"  ROC-AUC:   {val_results[best_model_name]['val_roc_auc']:.4f}")
    print(f"  Accuracy:  {val_results[best_model_name]['val_accuracy']:.4f}")
    print(f"  Precision: {val_results[best_model_name]['val_precision']:.4f}")
    print(f"  Recall:    {val_results[best_model_name]['val_recall']:.4f}")
    print(f"  F1-Score:  {val_results[best_model_name]['val_f1']:.4f}")
    
    print(f"\nTest Set (Final Performance):")
    print(f"  ROC-AUC:   {test_results[best_model_name]['roc_auc']:.4f}")
    print(f"  Accuracy:  {test_results[best_model_name]['accuracy']:.4f}")
    print(f"  Precision: {test_results[best_model_name]['precision']:.4f}")
    print(f"  Recall:    {test_results[best_model_name]['recall']:.4f}")
    print(f"  F1-Score:  {test_results[best_model_name]['f1']:.4f}")
    
    return best_model_name

# ============================================================================
# STEP 7: SAVE RESULTS
# ============================================================================

def save_experiment_results(models, val_results, test_results, preprocessor, 
                           best_model_name, feature_names):
    """Save all experiment artifacts"""
    
    print("\n" + "="*80)
    print("STEP 7: SAVING EXPERIMENT RESULTS")
    print("="*80)
    
    import os
    os.makedirs(Config.MODEL_DIR, exist_ok=True)
    os.makedirs(Config.OUTPUT_DIR, exist_ok=True)
    
    timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
    
    # Save best model
    best_model = models[best_model_name]
    model_path = f"{Config.MODEL_DIR}/best_model_{timestamp}.pkl"
    with open(model_path, 'wb') as f:
        pickle.dump(best_model, f)
    print(f"✅ Saved best model: {model_path}")
    
    # Save preprocessor
    preprocessor_path = f"{Config.MODEL_DIR}/preprocessor_{timestamp}.pkl"
    with open(preprocessor_path, 'wb') as f:
        pickle.dump(preprocessor, f)
    print(f"✅ Saved preprocessor: {preprocessor_path}")
    
    # Save experiment metadata
    metadata = {
        'experiment_name': Config.EXPERIMENT_NAME,
        'timestamp': timestamp,
        'best_model': best_model_name,
        'feature_strategy': Config.FEATURE_STRATEGY,
        'feature_names': feature_names,
        'data_splits': {
            'train': Config.TRAIN_SIZE,
            'val': Config.VAL_SIZE,
            'test': Config.TEST_SIZE
        },
        'validation_results': {
            name: {k: float(v) if isinstance(v, (np.float32, np.float64)) else v 
                   for k, v in results.items() if k not in ['val_predictions', 'val_probabilities']}
            for name, results in val_results.items()
        },
        'test_results': {
            name: {k: float(v) if isinstance(v, (np.float32, np.float64)) else v 
                   for k, v in results.items() if k not in ['predictions', 'probabilities']}
            for name, results in test_results.items()
        }
    }
    
    metadata_path = f"{Config.OUTPUT_DIR}/experiment_metadata_{timestamp}.json"
    with open(metadata_path, 'w') as f:
        json.dump(metadata, f, indent=2)
    print(f"✅ Saved metadata: {metadata_path}")
    
    print(f"\n📦 All artifacts saved!")
    print(f"   Model: {model_path}")
    print(f"   Preprocessor: {preprocessor_path}")
    print(f"   Metadata: {metadata_path}")
    
    return model_path, preprocessor_path, metadata_path

# ============================================================================
# MAIN PIPELINE
# ============================================================================

def main():
    """Complete end-to-end pipeline"""
    
    print("\n" + "="*80)
    print("🚀 PRODUCTION ML PIPELINE - EV EFFICIENCY PREDICTION")
    print("="*80)
    print(f"Experiment: {Config.EXPERIMENT_NAME}")
    print(f"Strategy: {Config.FEATURE_STRATEGY}")
    
    # Load data
    print(f"\nLoading data from: {Config.DATA_PATH}")
    df = pd.read_csv(Config.DATA_PATH)
    print(f"Dataset shape: {df.shape}")
    
    # STEP 1: Split data BEFORE any preprocessing
    train_df, val_df, test_df = split_data_before_preprocessing(df)
    
    # STEP 2: Feature engineering (same for all splits, no leakage)
    print("\n" + "="*80)
    print("STEP 2: FEATURE ENGINEERING")
    print("="*80)
    print("Creating derived features...")
    
    train_df = engineer_features(train_df)
    val_df = engineer_features(val_df)
    test_df = engineer_features(test_df)
    print("✅ Features engineered for all splits")
    
    # STEP 3: Feature selection
    numerical_features, categorical_features = select_features(Config.FEATURE_STRATEGY)
    feature_names = numerical_features + categorical_features
    
    # STEP 4: Preprocessing (fit on train, transform all)
    preprocessor = DataPreprocessor(numerical_features, categorical_features)
    preprocessor.fit(train_df)
    
    X_train, y_train = preprocessor.transform(train_df, 'Train')
    X_val, y_val = preprocessor.transform(val_df, 'Validation')
    X_test, y_test = preprocessor.transform(test_df, 'Test')
    
    # STEP 5: Train models
    trainer = ModelTrainer()
    models, val_results = trainer.train_all_models(X_train, y_train, X_val, y_val)
    
    # STEP 6: Final test evaluation
    test_results = trainer.evaluate_on_test(X_test, y_test)
    
    # STEP 7: Compare and select best model
    best_model_name = compare_and_select_model(val_results, test_results, y_val)
    
    # STEP 8: Save everything
    model_path, preprocessor_path, metadata_path = save_experiment_results(
        models, val_results, test_results, preprocessor, 
        best_model_name, feature_names
    )
    
    # Final summary
    print("\n" + "="*80)
    print("✅ PIPELINE COMPLETED SUCCESSFULLY!")
    print("="*80)
    
    print(f"\n🎯 KEY TAKEAWAYS:")
    print(f"  1. Best Model: {best_model_name}")
    print(f"  2. Validation ROC-AUC: {val_results[best_model_name]['val_roc_auc']:.4f}")
    print(f"  3. Test ROC-AUC: {test_results[best_model_name]['roc_auc']:.4f}")
    print(f"  4. Overfitting Gap: {val_results[best_model_name]['overfitting_gap']:.4f}")
    
    print(f"\n💡 NEXT STEPS FOR TEAM:")
    print(f"  1. Review validation vs test performance")
    print(f"  2. Check if overfitting is acceptable (< 0.05 is good)")
    print(f"  3. Share model artifacts: {Config.MODEL_DIR}/")
    print(f"  4. Document in team wiki/confluence")
    print(f"  5. Proceed to deployment planning")
    
    return models, val_results, test_results, preprocessor, best_model_name

if __name__ == "__main__":
    models, val_results, test_results, preprocessor, best_model_name = main()
```
